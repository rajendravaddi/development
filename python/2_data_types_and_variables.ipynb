# Variables

A **variable** is a name used to refer to a value stored in a program. 

Variables allow you to store, access, and manipulate data throughout your code.

## Creating Variables

Python does not require you to declare variables explicitly. 

A variable is created automatically when you assign a value to it using the assignment operator (`=`).


```python
x = 5
y = "John"

print(x)
print(y)
```

### Changing a Variable's Value and Type

Python is dynamically typed, so you do not need to specify a variable's type when creating it. 

You can also assign a value of a different type to the same variable.


```python
x = 4        # x refers to an integer
x = "Sally"  # x now refers to a string

print(x)    # The variable `x` initially refers to the integer `4`. After the second assignment, it refers to the string `"Sally"`.
```

## Rules for Naming Variables

Python variable names must follow these rules:

- A variable name must begin with a letter (`A–Z` or `a–z`) or an underscore (`_`).
- It can contain letters, digits (`0–9`), and underscores.
- It cannot begin with a digit.
- It cannot contain spaces or special characters such as `@`, `$`, or `%`.
- Variable names are case-sensitive.
- Python keywords, such as `if`, `for`, and `class`, cannot be used as variable names.


```python
name = "John"       # Valid
_age = 25           # Valid
student1 = "Alice"  # Valid
total_marks = 95    # Valid
```

## Global Variables

A global variable is a variable defined outside all functions. 

It can generally be accessed both inside and outside functions, provided it is in scope.

```python
x = "awesome"  # Global variable

def myfunc():
    print("Python is " + x)

myfunc()
```

## Local Variables and Global Variables with the Same Name

If you create a variable inside a function with the same name as a global variable, Python treats it as a separate local variable.

The local variable does not overwrite the global variable.

```python
x = "awesome"  # Global variable

def myfunc():
    x = "fantastic"  # Local variable
    print("Python is " + x)

myfunc()

print("Python is " + x)
```

## The global Keyword

Normally, assigning a value to a variable inside a function makes that name local to the function.

The global keyword tells Python that a variable name refers to the variable in the global scope rather than creating a local variable.

```python
def myfunc():
    global x
    x = "fantastic"

myfunc()

print("Python is " + x)
```

```python
x = "awesome"

def myfunc():
    global x
    x = "fantastic"

myfunc()

print("Python is " + x)
```

# Data Types

a **data type** defines the kind of value an object represents and the operations that can be performed on it.

Variables can store values of different data types, such as text, numbers, collections, and Boolean values. 

Python is dynamically typed, which means you do not need to declare a variable's data type explicitly. 

Python determines the type based on the value assigned to it.

## Built-in Data Types

Python provides several built-in data types, organized into the following categories:

| Category | Data Types | Description |
|---|---|---|
| Text Type | `str` | Represents a sequence of characters. |
| Numeric Types | `int`, `float`, `complex` | Represents integers, decimal numbers, and complex numbers. |
| Sequence Types | `list`, `tuple`, `range` | Represents ordered collections of items or sequences of numbers. |
| Mapping Type | `dict` | Stores data as key-value pairs. |
| Set Types | `set`, `frozenset` | Stores unique elements. |
| Boolean Type | `bool` | Represents `True` or `False`. |
| Binary Types | `bytes`, `bytearray`, `memoryview` | Represents or provides access to binary data. |
| None Type | `NoneType` | Represents the absence of a value using `None`. |

## Getting the Data Type

You can use Python's built-in `type()` function to determine the type of an object.

```python
print(type("Hello"))  # <class 'str'>
print(type(10))       # <class 'int'>
print(type(3.14))     # <class 'float'>
print(type(True))     # <class 'bool'>
print(type(None))     # <class 'NoneType'>
```



## Assigning Values to Variables

In Python, the data type of a variable is determined by the value assigned to it. You can assign values of different types without declaring their types in advance.

The following table shows examples of Python variables and their corresponding data types.

| Example | Data Type |
|---|---|
| `x = "Hello World"` | `str` |
| `x = 20` | `int` |
| `x = 20.5` | `float` |
| `x = 1j` | `complex` |
| `x = ["apple", "banana", "cherry"]` | `list` |
| `x = ("apple", "banana", "cherry")` | `tuple` |
| `x = range(6)` | `range` |
| `x = {"name": "John", "age": 36}` | `dict` |
| `x = {"apple", "banana", "cherry"}` | `set` |
| `x = frozenset({"apple", "banana", "cherry"})` | `frozenset` |
| `x = True` | `bool` |
| `x = b"Hello"` | `bytes` |
| `x = bytearray(5)` | `bytearray` |
| `x = memoryview(bytes(5))` | `memoryview` |
| `x = None` | `NoneType` |


Here, the same variable `x` first stores an integer and then stores a string. Python allows this because it is dynamically typed.


```python
a = 256
b = 256
print(a == b)
print(a is b)
```

```python
x = 257
y = 257
print(x == y)
print(x is y)
```

```python
a1 = -5
a2 = -5
a1 is a2
```

```python
a3 = -6
a4 = -6
a3 is a4
```

CPython traditionally caches integers from -5 to 256, so it commonly reuses the same objects for integers -5 to 256. The integers 257 and -6 are outside that traditional cache range, so separate objects may be used.

```python
b1 = True
b2 = True
b1 is b2
```


## Specifying a Data Type Using Constructor Functions

Python provides built-in constructor functions that can create objects of specific data types. These functions can also convert compatible values from one type to another.

| Constructor | Example | Resulting Type |
|---|---|---|
| `str()` | `str(123)` | `str` |
| `int()` | `int(20.8)` | `int` |
| `float()` | `float(20)` | `float` |
| `complex()` | `complex(1, 2)` | `complex` |
| `list()` | `list(("apple", "banana"))` | `list` |
| `tuple()` | `tuple(["apple", "banana"])` | `tuple` |
| `range()` | `range(6)` | `range` |
| `dict()` | `dict(name="John", age=36)` | `dict` |
| `set()` | `set(("apple", "banana"))` | `set` |
| `frozenset()` | `frozenset(("apple", "banana"))` | `frozenset` |
| `bool()` | `bool(1)` | `bool` |
| `bytes()` | `bytes(5)` | `bytes` |
| `bytearray()` | `bytearray(5)` | `bytearray` |
| `memoryview()` | `memoryview(bytes(5))` | `memoryview` |
