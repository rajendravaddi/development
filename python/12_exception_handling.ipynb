# Python Exception Handling

Exception handling in Python is a robust mechanism used to detect, intercept, and gracefully handle runtime errors without abruptly crashing the application.


### 1. What is an Exception?

**Explanation**: An exception is a special signal or event triggered during program execution that disrupts the normal flow of instructions. 

If unhandled, Python halts execution and outputs a traceback indicating where and why the error occurred.


````python
# Unhandled exception causes immediate program termination
a = 10
b = 0
result = a / b
````

````python
# When an unhandled error occurs, statements after it will never execute.
try:
    a = 10
    b = 0
    print("Attempting division...")
    result = a / b
    print("Result:", result)  # Skipped!
except ZeroDivisionError as err:
    print(f"Caught an exception: {err}")

print("Program continues smoothly without crashing!")
````

## 2. Types of Errors in Python: Syntax Errors vs Runtime Exceptions

Python errors generally fall into two primary categories:

### A. Syntax Errors (Parsing Errors)
Occur when the Python interpreter parses source code and encounters code that violates Python grammatical rules (e.g., missing colons, unmatched parentheses). 

Syntax errors occur **before** the program begins execution and **cannot** be caught by a `try-except` block.


````python

# SyntaxError: cannot be handled at runtime
if 10 > 5
    print("Missing colon")
````



### B. Runtime Errors (Exceptions)
Occur while the program is actively running. 

The syntax is 100% valid, but an invalid operation is attempted (e.g., dividing by zero, accessing an out-of-bounds list index, reading a missing file).

### Common Built-in Exceptions Hierarchy:
```text
BaseException
 ├── SystemExit
 ├── KeyboardInterrupt
 ├── GeneratorExit
 └── Exception  <-- (Base class for all standard application exceptions)
      ├── ArithmeticError (ZeroDivisionError, OverflowError)
      ├── LookupError (IndexError, KeyError)
      ├── TypeError
      ├── ValueError
      ├── NameError
      ├── AttributeError
      └── OSError (FileNotFoundError, PermissionError)
```

| Exception | Explanation |
| :--- | :--- |
| `ZeroDivisionError` | Dividing a number by zero (`10 / 0`) |
| `TypeError` | Applying an operation or function to an inappropriate type (`'5' + 5`) |
| `ValueError` | Argument has right type but inappropriate value (`int('abc')`) |
| `IndexError` | Accessing a sequence index out of bounds (`[1, 2][5]`) |
| `KeyError` | Looking up a non-existent key in a dictionary (`{'a': 1}['b']`) |
| `NameError` | Referencing an identifier that has not been defined |
| `AttributeError` | Accessing an attribute/method that does not exist on the object |
| `FileNotFoundError` | Attempting to open a file path that does not exist |
| `ModuleNotFoundError`| Importing a module that is not installed or found |

````python
# 1. IndexError
try:
    numbers = [10, 20, 30]
    print(numbers[5])
except IndexError as e:
    print("1. IndexError caught       :", e)
````

````python
# 2. KeyError
try:
    user = {"name": "Alice"}
    print(user["email"])
except KeyError as e:
    print("2. KeyError caught         :", e)
````

````python
# 3. TypeError
try:
    total = "Score: " + 100
except TypeError as e:
    print("3. TypeError caught        :", e)
````

````python
# 4. ValueError
try:
    val = int("not_a_number")
except ValueError as e:
    print("4. ValueError caught       :", e)
````

### 3. The Core `try` and `except` Blocks

**Explanation**: The fundamental construct for exception handling. Code that could potentially fail is placed inside `try`. If an exception occurs, execution immediately jumps to the matching `except` block.

**Syntax**:
```python
try:
    # Protected / risky code
    ...
except SpecificException:
    # Recovery or fallback code
    ...
```

````python
# 3. try and except execution flow
print("Step 1: Before try block")

try:
    print("Step 2: Entering try block")
    a = 10
    b = 0
    res = a / b
    print("Step 3: This line will NOT execute!")
except ZeroDivisionError:
    print("Step 4: Exception intercepted! Handling gracefully.")

print("Step 5: Execution resumes after try-except block.")
````

### 4. Multiple Specific `except` Blocks

**Explanation**: A single `try` block can have multiple `except` blocks to handle distinct exceptions with specialized recovery logic. Python checks handlers from top to bottom and executes ONLY the first matching block.

**Syntax**:
```python
try:
    # risky code
except ZeroDivisionError:
    # handle division by zero
except ValueError:
    # handle invalid conversion
except TypeError:
    # handle invalid type
```

````python
# 4. Multiple except blocks demonstration
def safe_calculator(val1, val2):
    try:
        num1 = int(val1)
        num2 = int(val2)
        result = num1 / num2
        print(f"Success: {num1} / {num2} = {result}")
    except ZeroDivisionError:
        print(f"Math Error: Cannot divide by zero! (Denominator was {val2})")
    except ValueError:
        print(f"Input Error: Please provide valid integers! (Got '{val1}', '{val2}')")
    except TypeError:
        print("Type Error: Incompatible argument types passed!")

print("="*25)
safe_calculator(10, 2)     # Success
print("="*25)
safe_calculator(10, 0)     # ZeroDivisionError
print("="*25)
safe_calculator("abc", 5)  # ValueError
````

### 5. Catching Multiple Exceptions in a Single `except` Block

**Explanation**: When several distinct exception types require the exact same recovery or cleanup logic, they can be grouped together as a tuple inside a single `except` statement.

**Syntax**:
```python
try:
    # risky code
except (ValueError, ZeroDivisionError, TypeError) as err:
    # shared recovery logic
```

````python
# 5. Handling multiple exceptions in a single tuple
def parse_and_divide(input_a, input_b):
    try:
        a = int(input_a)
        b = int(input_b)
        return a / b
    except (ValueError, ZeroDivisionError) as err:
        print(f"[Warning] Handled ({type(err).__name__}): {err}")
        return None

print("Result 1:", parse_and_divide("10", "2"))
print("Result 2:", parse_and_divide("10", "0"))      # ZeroDivisionError caught
print("Result 3:", parse_and_divide("invalid", "2")) # ValueError caught
````

### 6. Generic Exception Handling (`Exception as e`) & Best Practices

**Explanation**: Catching `Exception` intercepts all standard runtime exceptions (since virtually all built-in exceptions inherit from `Exception`). Using `as e` assigns the exception instance to inspect its message and class name.

**Syntax**:
```python
try:
    # risky code
except Exception as e:
    print(f'Error occurred: {e}')
```

````python
# 6. Generic Exception Catching and Best Practice
# BEST PRACTICE:
# 1. Catch specific exceptions first.
# 2. Keep generic 'Exception' at the very bottom as a fallback.
# 3. Avoid bare 'except:' because it catches KeyboardInterrupt & SystemExit!

def process_data(data, key, divisor):
    try:
        val = data[key]
        return val / divisor
    except KeyError:
        print(f"Key Error: Key '{key}' not found in dictionary.")
    except ZeroDivisionError:
        print("Math Error: Division by zero attempted.")
    except Exception as e:
        # Fallback catch-all for unexpected issues
        print(f"Unexpected error ({type(e).__name__}): {e}")

sample = {"count": 100}
process_data(sample, "count", 0)       # Specific ZeroDivisionError
process_data(sample, "missing", 5)     # Specific KeyError
process_data(sample, "count", "bad")   # Caught by generic fallback (TypeError)
````

### 7. The `else` Block (Success Path)

**Explanation**: The `else` block executes ONLY IF the `try` block completed without raising any exceptions. It cleanly separates the protected risky operation from the logic that depends on its success.

**Syntax**:
```python
try:
    # risky operation ONLY
except SomeError:
    # handle error
else:
    # executes ONLY when try succeeded without errors
```

````python
# 7. else block example
def calculate_ratio(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Denominator cannot be zero.")
    else:
        # Executes only if division was successful
        print(f"Success! The calculated ratio is {result:.4f}")
        return result

print("--- Scenario 1: Success ---")
calculate_ratio(15, 4)

print("\n--- Scenario 2: Error ---")
calculate_ratio(15, 0)
````

### 8. The `finally` Block (Guaranteed Cleanup)

**Explanation**: The `finally` block ALWAYS executes before leaving the `try-except` statement, regardless of whether an exception occurred, whether it was caught, or whether an early `return` was called. Ideal for closing files, network connections, or database sessions.

**Syntax**:
```python
try:
    # allocate resource
except SomeError:
    # handle error
finally:
    # ALWAYS executed (release resource)
```

````python
# 8. finally block execution guarantee
def divide_with_cleanup(x, y):
    try:
        print("1. [try] Opening resources and computing...")
        res = x / y
        print(f"2. [try] Computation succeeded: {res}")
        return res
    except ZeroDivisionError:
        print("2. [except] Handling division by zero error.")
        return 0
    finally:
        # This runs EVEN AFTER the return statements above!
        print("3. [finally] Releasing and cleaning up resources.")

print("Call 1 (Success):")
val1 = divide_with_cleanup(10, 2)
print("Returned:", val1)

print("\nCall 2 (Error):")
val2 = divide_with_cleanup(10, 0)
print("Returned:", val2)
````

---
## 9. The Complete `try-except-else-finally` Lifecycle

When combining all four blocks:
* `try`: Runs the initial risky code.
* `except`: Runs **only if** an exception occurred.
* `else`: Runs **only if** NO exception occurred.
* `finally`: Runs **always** under all circumstances before moving on.

```text
       ┌──────────────┐
       │  try block   │
       └──────┬───────┘
              │
       Exception raised?
        /            \
      YES             NO
      ▼               ▼
┌──────────────┐ ┌──────────────┐
│ except block │ │  else block  │
└──────┬───────┘ └──────┬───────┘
       │                │
       └───────┬────────┘
               ▼
       ┌──────────────┐
       │finally block │  (Always runs!)
       └──────────────┘
```

````python
# 9. Complete try-except-else-finally workflow
def execute_transaction(amount, divisor):
    print(f"\n--- Executing Transaction (amount={amount}, divisor={divisor}) ---")
    try:
        print("  [try] Computing transaction amount...")
        result = amount / divisor
    except ZeroDivisionError:
        print("  [except] Error: Cannot divide transaction by zero!")
    except TypeError as e:
        print(f"  [except] Type Error: {e}")
    else:
        print(f"  [else] Transaction approved! Dispatched: {result}")
    finally:
        print("  [finally] Logging transaction audit trail to database.")

execute_transaction(500, 5)   # try -> else -> finally
execute_transaction(500, 0)   # try -> except -> finally
````

### 10. The `raise` Keyword: Explicitly Triggering Exceptions

**Explanation**: Used to intentionally trigger an exception when program constraints, validation rules, or business logic are violated.

**Syntax**:
```python
if condition_invalid:
    raise ValueError('Descriptive error message')
```

````python
# 10. Explicitly raising exceptions
def register_voter(name: str, age: int):
    if not isinstance(age, int):
        raise TypeError(f"Age must be an integer, got {type(age).__name__}")
    if age < 0 or age > 130:
        raise ValueError(f"Invalid age provided: {age}. Must be between 0 and 130.")
    if age < 18:
        raise PermissionError(f"{name} is {age} years old and ineligible to vote.")
    
    print(f"Voter '{name}' registered successfully.")

# Valid registration
register_voter("Alice", 25)

# Triggering validation errors
try:
    register_voter("Bob", 16)
except PermissionError as err:
    print("Caught validation error:", err)

try:
    register_voter("Charlie", -5)
except ValueError as err:
    print("Caught validation error:", err)
````

### 11. Custom Exceptions (User-Defined Exceptions)

**Explanation**: Custom exceptions represent domain-specific error conditions in your application (e.g., banking, authentication, commerce). Custom exceptions inherit from `Exception` (or a subclass of it).

**Syntax**:
```python
class MyCustomError(Exception):
    """Docstring explaining error."""
    pass
```

````python
# 11. Custom domain-specific exception hierarchy
class BankingError(Exception):
    """Base exception class for all banking errors."""
    pass

class InsufficientBalanceError(BankingError):
    """Raised when withdrawal exceeds available funds."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Cannot withdraw ${amount:,.2f}; current balance is ${balance:,.2f} (short by ${amount - balance:,.2f})")

class AccountLockedError(BankingError):
    """Raised when attempting operations on a frozen/locked account."""
    pass

# Account class using custom exceptions
class BankAccount:
    def __init__(self, account_id: str, balance: float, is_locked: bool = False):
        self.account_id = account_id
        self.balance = balance
        self.is_locked = is_locked

    def withdraw(self, amount: float):
        if self.is_locked:
            raise AccountLockedError(f"Account '{self.account_id}' is locked. Contact support.")
        if amount > self.balance:
            raise InsufficientBalanceError(self.balance, amount)
        self.balance -= amount
        print(f"Withdrawal of ${amount:.2f} approved. Remaining: ${self.balance:.2f}")

# Testing custom exceptions
account = BankAccount(account_id="ACC-9874", balance=5000.0)

try:
    account.withdraw(7500.0)
except InsufficientBalanceError as e:
    print("Banking Alert:", e)
    print(f"Requested: ${e.amount}, Available: ${e.balance}")
````

### 12. Exception Propagation & Call Stack Unwinding

**Explanation**: When an exception occurs inside a nested function, Python searches the current function for a handler. If none is found, the current function exits immediately, and Python unwinds the call stack to the caller, repeating until a handler is found or the program terminates.

**Syntax**:
```python
# func1() calls func2() -> calls func3() -> raises exception
# Caught at top level in func1()
```

````python
# 12. Exception propagation through the call stack
def step_three():
    print("  Entering step_three()")
    # Root of the exception
    result = 10 / 0
    print("  Leaving step_three()")

def step_two():
    print(" Entering step_two()")
    step_three()
    print(" Leaving step_two()")

def step_one():
    print("Entering step_one()")
    step_two()
    print("Leaving step_one()")

try:
    print("--- Starting Call Chain ---")
    step_one()
except ZeroDivisionError as err:
    print("\n[TOP LEVEL HANDLER] Caught propagated error:", err)
````

### 13. Re-raising Exceptions (`raise`)

**Explanation**: Allows an `except` block to intercept an error, perform logging or partial cleanup, and then re-raise the exact same exception so that outer calling functions remain aware of the failure.

**Syntax**:
```python
try:
    ...
except SomeError as err:
    log_error(err)
    raise  # Re-raises the active exception unchanged
```

````python
# 13. Re-raising an intercepted exception
def fetch_user_data(user_id):
    try:
        # Simulate network or database query
        if user_id <= 0:
            raise ValueError(f"Invalid user_id {user_id}: ID must be a positive integer.")
        return {"id": user_id, "name": "Alice"}
    except ValueError as e:
        print(f"[AUDIT LOG] Logged error at service level: {e}")
        # Re-raise so the caller can decide how to handle it
        raise

# Top-level application handling the re-raised error
try:
    user = fetch_user_data(-1)
except ValueError as e:
    print(f"[UI LAYER] Displaying user-friendly alert: {e}")
````

### 14. Exception Chaining (`raise ... from ...`)

**Explanation**: PEP 3134 introduced exception chaining to wrap lower-level exceptions into higher-level domain exceptions while preserving the root cause traceback via `__cause__`.

**Syntax**:
```python
try:
    low_level_call()
except LowLevelError as e:
    raise HighLevelDomainError('Contextual failure') from e
```

````python
# 14. Exception chaining with 'from'
class DatabaseConnectionError(Exception):
    """Custom high-level application exception."""
    pass

def load_system_configuration(config_file):
    try:
        # Low-level error: File does not exist
        with open(config_file, "r") as f:
            return f.read()
    except FileNotFoundError as original_err:
        # Chain into application-level error
        raise DatabaseConnectionError("Failed to initialize system: configuration source unreachable.") from original_err

try:
    load_system_configuration("non_existent_config.json")
except DatabaseConnectionError as err:
    print("High-level error:", err)
    print("Original root cause (__cause__):", repr(err.__cause__))
````

### 15. The `assert` Statement

**Explanation**: Checks internal invariants and assumptions during development and testing. If the condition is False, Python raises an `AssertionError`. Note: Assertions can be disabled in production using Python's `-O` (optimize) flag, so never use them for untrusted user input validation!

**Syntax**:
```python
assert condition, 'Assertion failure error message'
```

````python
# 15. Assertions for internal consistency
def calculate_discount(price: float, discount_rate: float) -> float:
    # Internal assumption assertions:
    assert price >= 0, f"Price must be non-negative, got {price}"
    assert 0 <= discount_rate <= 1.0, f"Discount rate must be between 0 and 1, got {discount_rate}"
    
    return price * (1.0 - discount_rate)

print("Valid discount :", calculate_discount(100.0, 0.20))

try:
    # Violating internal assertion
    calculate_discount(100.0, 1.50)
except AssertionError as e:
    print("Assertion caught:", e)
````

### 16. Exception Handling in File Operations & Context Managers (`with`)

**Explanation**: Working with external I/O (files, sockets) requires defensive exception handling. While `try-finally` works, Python's `with` statement (Context Manager protocol) automatically guarantees file closure even if exceptions occur.

**Syntax**:
```python
# Preferred: Context manager handles automatic file closure
try:
    with open('file.txt', 'r') as f:
        content = f.read()
except (FileNotFoundError, PermissionError) as err:
    ...
```

````python
# 16. Comparing manual file handling vs context manager
import os

filename = "temp_demo_log.txt"

# Clean up any leftover file
if os.path.exists(filename):
    os.remove(filename)

# 1. Approach 1: Classic try-finally (Verbose)
try:
    f = open(filename, "w")
    f.write("Line 1: Log entry\n")
finally:
    f.close()
    print("Approach 1: File explicitly closed via finally.")

# 2. Approach 2: Modern with statement (Idiomatic & Safe)
try:
    with open(filename, "r") as f:
        content = f.read()
        print("Approach 2 Content:\n", content.strip())
except FileNotFoundError:
    print("File was not found.")
except PermissionError:
    print("Permission denied when accessing file.")

# Clean up
if os.path.exists(filename):
    os.remove(filename)
    print("Temporary log file cleaned up.")
````
