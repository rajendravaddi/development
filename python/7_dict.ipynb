## 1. Definition and Creation

A **dictionary** (`dict`) is a mutable mapping of unique **keys** to arbitrary **values**.
* **Key-Value Pairs**: Associative mappings.
* **Insertion Ordered**: Since Python 3.7+, dictionaries are guaranteed to maintain insertion order.
* **Fast Lookup**: Constant average time $O(1)$ key lookup, insertion, and deletion.

### Creation Ways:
* Curly braces: `{}` or `{"a": 1, "b": 2}`
* Constructor: `dict()`, `dict(a=1, b=2)`, or `dict([('a', 1), ('b', 2)])`
* `dict.fromkeys(keys, default)`
* Dictionary comprehensions

````python
# 1. Dictionary Creation Examples
empty_dict = {}
dict_constructor = dict()

# Literal syntax
user = {
    "name": "Alice",
    "age": 30,
    "is_active": True,
    "skills": ["Python", "SQL"]
}

# Constructor with keyword arguments
config = dict(host="localhost", port=8080, debug=True)

# Constructor with list of key-value tuples
from_pairs = dict([("id", 101), ("status", "ok")])

print("Literal dict :", user)
print("Config dict  :", config)
print("From pairs   :", from_pairs)
````

## 2. Hash Tables & Memory Management (Compact Dictionaries)

### How Python Dictionaries Work Under the Hood:
1. **Hash Table**:
   When you write `d[key]`, Python computes `hash(key)`. The hash value determines the index in an internal table.
2. **Compact Dict Architecture (PEP 468 / Python 3.6+)**:
   Previously, dictionaries stored entries in a sparse array holding `(hash, key, value)`, wasting 30-40% of memory on empty slots.
   Modern Python uses:
   * **Sparse indices array**: Small table mapping hash indices to entry row numbers.
   * **Dense entries array**: Sequential table storing `[hash, key, value]` in exact insertion order!
   This cut memory usage by ~30% and made dictionaries **natively insertion-ordered**.

````python
# 2. Hash Table Inspection
key1 = "apple"
key2 = 12345
key3 = (1, 2)

print(f"hash('{key1}') = {hash(key1)}")
print(f"hash({key2})   = {hash(key2)}")
print(f"hash({key3})   = {hash(key3)}")

# Insertion order preservation (Python 3.7+)
ordered_demo = {}
ordered_demo["z"] = 1
ordered_demo["a"] = 2
ordered_demo["m"] = 3

print("\nKeys maintain insertion order:")
print(list(ordered_demo.keys()))
````

## 3. Key Requirements: Hashability

A dictionary key **must be hashable**:
1. An object is hashable if it has a hash value that never changes during its lifetime (`__hash__()`).
2. It can be compared to other objects for equality (`__eq__()`).
3. Atomic immutable types (`int`, `float`, `str`, `bytes`) and tuples containing only hashable objects are hashable.
4. Mutable types (`list`, `dict`, `set`) are **not hashable** and cannot be dictionary keys.

````python
# 3. Hashability Examples
valid_keys = {
    42: "integer key",
    3.14: "float key",
    "user_id": "string key",
    (10, 20): "tuple key",
    frozenset([1, 2]): "frozenset key"
}
print("Valid hashable keys dictionary:", valid_keys)

# Attempting to use a mutable list as key
try:
    bad_dict = {[1, 2]: "error"}
except TypeError as e:
    print("\nCaught expected error using list key:", e)
````

## 4. Accessing, Modifying & Deleting Items

* **Access via `[]`**: `d[key]` (raises `KeyError` if key is not present).
* **Safe access via `.get()`**: `d.get(key, default)` (returns `default` if missing).
* **Adding / Updating**: `d[key] = val`
* **Deleting**: `del d[key]` or `d.pop(key)`

````python
# 4. Access, Update & Delete Example
book = {"title": "Fluent Python", "pages": 800, "author": "Luciano Ramalho"}

# 1. Access
print("Title:", book["title"])

# Missing key access
try:
    print(book["publisher"])
except KeyError as e:
    print("Caught KeyError for:", e)

# Safe access with get()
print("Publisher (safe):", book.get("publisher", "Unknown Publisher"))

# 2. Add / Update
book["year"] = 2022        # Add new key
book["pages"] = 1012       # Update existing key
print("\nAfter update:", book)

# 3. Delete
del book["year"]
print("After del 'year':", book)
````

## 5. Dictionary Views (`keys()`, `values()`, `items()`)

`d.keys()`, `d.values()`, and `d.items()` return **view objects**.
* View objects do **not** duplicate data in memory; they are dynamic windows reflecting changes made to the dictionary in real-time.
* `dict_keys` and `dict_items` support set operations (`&`, `|`, `-`, `^`).

````python
# 5. Dynamic Dictionary Views Example
scores = {"Alice": 90, "Bob": 85}

k_view = scores.keys()
v_view = scores.values()
i_view = scores.items()

print("Initial keys view:", k_view)

# Mutating dictionary reflects immediately in the view!
scores["Charlie"] = 95
print("After adding Charlie:")
print("Keys view  :", k_view)
print("Values view:", v_view)
print("Items view :", i_view)

# Set operations on dict_keys
dict_a = {"x": 1, "y": 2, "z": 3}
dict_b = {"y": 20, "z": 30, "w": 40}

common_keys = dict_a.keys() & dict_b.keys()
print("\nCommon keys (set intersection):", common_keys)
````

## 6. Dictionary Comprehensions

Syntax:
```python
{key_expr: value_expr for item in iterable if condition}
```

````python
# 6. Dictionary Comprehension Examples
# Squares dictionary
squares = {x: x ** 2 for x in range(1, 6)}
print("Squares:", squares)

# Inverting / swapping keys and values
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print("Inverted:", inverted)

# Filtering with if condition
salaries = {"Alice": 90000, "Bob": 45000, "Charlie": 120000, "David": 55000}
high_earners = {name: sal for name, sal in salaries.items() if sal >= 60000}
print("High earners:", high_earners)
````

## 7. Modern Merge (`|`) and Update (`|=`) Operators (Python 3.9+)

Python 3.9 introduced:
* **Merge (`|`)**: Returns a new dictionary merging two dicts.
* **In-place Update (`|=`)**: Updates the left dictionary in place.

````python
# 7. Modern Dictionary Operators Example
defaults = {"theme": "light", "font": "Arial", "zoom": 100}
custom = {"theme": "dark", "zoom": 125}

# 1. Merge (|)
merged = defaults | custom  # custom values take precedence
print("Merged (|)          :", merged)
print("Defaults unchanged  :", defaults)

# 2. Update in-place (|=)
defaults |= custom
print("Defaults after |=   :", defaults)
````

## 8. All Built-in Dictionary Methods (One by One)

Python provides 11 built-in methods on `dict`. Each is detailed below with its functionality, syntax, and an executable code cell.

### `get()`

**Functionality**: Returns the value for key if key is in the dictionary; otherwise returns default (or None if default is not provided). Never raises KeyError.

**Syntax**:
```python
dict.get(key[, default=None]) -> value
```

````python
# get() example
user = {"name": "Bob", "role": "editor"}

print("Name        :", user.get("name"))
print("Email       :", user.get("email"))                 # None
print("Email (def) :", user.get("email", "no@email.com")) # Custom default
````

### `keys()`

**Functionality**: Returns a dynamic view object displaying all keys present in the dictionary.

**Syntax**:
```python
dict.keys() -> dict_keys
```

````python
# keys() example
inventory = {"apples": 10, "bananas": 5, "oranges": 8}
k = inventory.keys()

print("Keys view:", k)
print("Keys as list:", list(k))
````

### `values()`

**Functionality**: Returns a dynamic view object displaying all values present in the dictionary.

**Syntax**:
```python
dict.values() -> dict_values
```

````python
# values() example
inventory = {"apples": 10, "bananas": 5, "oranges": 8}
v = inventory.values()

print("Values view:", v)
print("Total sum of items:", sum(v))
````

### `items()`

**Functionality**: Returns a dynamic view object displaying all (key, value) tuple pairs in the dictionary.

**Syntax**:
```python
dict.items() -> dict_items
```

````python
# items() example
inventory = {"apples": 10, "bananas": 5, "oranges": 8}

for fruit, count in inventory.items():
    print(f"We have {count} {fruit}.")
````

### `pop()`

**Functionality**: Removes specified key and returns its value. If key is not found, returns default if given; otherwise raises KeyError.

**Syntax**:
```python
dict.pop(key[, default]) -> value
```

````python
# pop() example
user = {"id": 1, "username": "alice", "temp_token": "abc123xyz"}

token = user.pop("temp_token")
print(f"Removed token: {token}")
print("Dictionary now:", user)

# Safe pop with default
missing = user.pop("non_existent_key", "DEFAULT")
print("Missing key safe pop:", missing)
````

### `popitem()`

**Functionality**: Removes and returns the last (key, value) pair inserted into the dictionary as a tuple (LIFO order). Raises KeyError if empty.

**Syntax**:
```python
dict.popitem() -> tuple[key, value]
```

````python
# popitem() example
cache = {"a": 1, "b": 2, "c": 3}

k, v = cache.popitem()
print(f"Popped last item: ({k}, {v})")
print("Cache now:", cache)
````

### `setdefault()`

**Functionality**: If key is in the dictionary, returns its value. If not, inserts key with a value of default and returns default.

**Syntax**:
```python
dict.setdefault(key[, default=None]) -> value
```

````python
# setdefault() example
profile = {"name": "Charlie"}

# Key already exists -> returns current value, does NOT overwrite
val1 = profile.setdefault("name", "Unknown")
print(f"setdefault existing key: {val1}")

# Key does not exist -> inserts key with default and returns it
val2 = profile.setdefault("country", "Canada")
print(f"setdefault missing key : {val2}")
print("Profile dict now       :", profile)
````

### `update()`

**Functionality**: Updates the dictionary with key/value pairs from other, overwriting existing keys. Accepts another dictionary or an iterable of key/value pairs.

**Syntax**:
```python
dict.update([other], **kwargs) -> None
```

````python
# update() example
settings = {"volume": 70, "mute": False}

# Update with another dictionary
settings.update({"volume": 85, "theme": "dark"})
print("After dict update:", settings)

# Update with keyword arguments
settings.update(mute=True, language="EN")
print("After kwargs update:", settings)
````

### `fromkeys()`

**Functionality**: Class method that creates a new dictionary with keys from iterable and values set to value (default None).

**Syntax**:
```python
dict.fromkeys(iterable[, value=None]) -> dict
```

````python
# fromkeys() example
fields = ["id", "name", "email"]

# Default value None
empty_form = dict.fromkeys(fields)
print("fromkeys default:", empty_form)

# Custom scalar default
zero_form = dict.fromkeys(fields, 0)
print("fromkeys with 0 :", zero_form)

# WARNING: mutable default object (e.g. list) is shared by all keys!
# Use comprehension instead if you need separate lists: {k: [] for k in fields}
````

### `clear()`

**Functionality**: Removes all items from the dictionary in-place, leaving it empty.

**Syntax**:
```python
dict.clear() -> None
```

````python
# clear() example
session = {"user_id": 42, "token": "xyz"}
print("Before clear:", session)

session.clear()
print("After clear :", session)
````

### `copy()`

**Functionality**: Returns a shallow copy of the dictionary.

**Syntax**:
```python
dict.copy() -> dict
```

````python
# copy() example
original = {"a": 1, "b": [10, 20]}
copied = original.copy()

copied["a"] = 99
print("Original a:", original["a"])  # 1 (unaffected)
print("Copied a  :", copied["a"])    # 99
````

## 9. Specialized Dictionaries (`collections` module)

Python's `collections` standard library module provides powerful dictionary extensions:
1. `defaultdict`: Automatically initializes missing keys with a default factory.
2. `Counter`: Tracks counts of hashable objects.
3. `OrderedDict`: Dictionary with order-sensitive operations (`move_to_end()`).

````python
# 9. Specialized Dictionaries Example
from collections import defaultdict, Counter

# 1. defaultdict: Grouping words by length
words = ["apple", "bat", "bar", "atom", "book", "code"]
by_length = defaultdict(list)
for w in words:
    by_length[len(w)].append(w)
print("defaultdict grouped by length:")
print(dict(by_length))

# 2. Counter: Frequency analysis
letter_counts = Counter("abracadabra")
print("\nCounter counts:", letter_counts)
print("Top 2 most common:", letter_counts.most_common(2))
````
