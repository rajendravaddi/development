### 1. The `while` Loop

**Explanation**: Repeatedly executes a code block as long as the test condition remains True. You must ensure the loop state updates toward the termination condition to prevent infinite loops.

**Syntax**:
```python
while condition:
    # loop body
    # state update
```

````python
# while loop example
count = 1

while count <= 5:
    print(f"Count is: {count}")
    count += 1

print("Loop finished!")
````

### 2. The `for` Loop

**Explanation**: Iterates over elements of any iterable (such as a string, list, tuple, dictionary, set, range, or file) sequentially.

**Syntax**:
```python
for item in iterable:
    # process item
```

````python
# for loop examples across different collections
languages = ["Python", "Rust", "Go", "TypeScript"]
print("--- Iterating over a list ---")
for lang in languages:
    print("Language:", lang)

print("\n--- Iterating over a string ---")
for ch in "Code":
    print("Char:", ch)

print("\n--- Iterating over dictionary (keys and values) ---")
user = {"name": "Alice", "role": "Dev", "country": "India"}
for key, val in user.items():
    print(f" {key}: {val}")
````

### 3. The `range()` Function

**Explanation**: Generates an immutable sequence of numbers. range() is memory efficient because it calculates values on demand rather than storing them in RAM.

**Syntax**:
```python
range(stop)
range(start, stop[, step])
```

````python
# range() variations
print("range(5)         :", list(range(5)))          # 0 to 4
print("range(2, 8)      :", list(range(2, 8)))       # 2 to 7
print("range(0, 20, 5)  :", list(range(0, 20, 5)))   # step 5: 0, 5, 10, 15
print("range(10, 0, -2) :", list(range(10, 0, -2)))  # countdown: 10, 8, 6, 4, 2

# Iterating with range:
total = 0
for i in range(1, 11):
    total += i
print(f"Sum of 1 to 10: {total}")
````

### 4. `break` Statement

**Explanation**: Immediately terminates the innermost loop, skipping any remaining code in the loop body and ignoring the loop's else block.

**Syntax**:
```python
for item in iterable:
    if condition:
        break
```

````python
# break example: search for an item and stop
numbers = [13, 27, 42, 59, 71, 88]
target = 42

for index, num in enumerate(numbers):
    print(f"Checking index {index}: {num}")
    if num == target:
        print(f"-> Found target {target} at index {index}! Breaking.")
        break
````

### 5. `continue` Statement

**Explanation**: Skips the rest of the current iteration's code body and immediately jumps to the next iteration of the loop.

**Syntax**:
```python
for item in iterable:
    if condition:
        continue
```

````python
# continue example: skip odd numbers
for num in range(1, 11):
    if num % 2 != 0:
        continue  # skip odds
    print(f"Even number: {num}")
````

### 6. `pass` Statement

**Explanation**: A null operation (no-op). Python syntax does not allow empty code blocks (under if, for, def, class); pass serves as a placeholder.

**Syntax**:
```python
for item in iterable:
    pass  # to be implemented later
```

````python
# pass example
for i in range(5):
    if i == 2:
        pass  # TODO: Handle special event for 2 in future release
    else:
        print(f"Processed item: {i}")

# Commonly used in empty function/class skeletons
class CustomError(Exception):
    pass
````

### 7. The Loop `else` Clause

**Explanation**: A unique Python feature: the else block attached to a for or while loop executes ONLY IF the loop completes normally without encountering a `break` statement.

**Syntax**:
```python
for item in iterable:
    if condition:
        break
else:
    # executed if NO break occurred
```

````python
# Loop else example: Prime number checker
def check_prime(n):
    if n < 2:
        print(f"{n} is not prime.")
        return
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            print(f"{n} is not prime (divisible by {i}).")
            break
    else:
        # Runs ONLY if the loop ran to completion without hitting break!
        print(f"{n} is a PRIME number!")

check_prime(17)
check_prime(15)
````

### 8. `enumerate()`: Tracking Index & Value

**Explanation**: Returns an iterator of tuples containing count index (starting from 0 or custom start) and the corresponding item from iterable.

**Syntax**:
```python
for index, item in enumerate(iterable[, start=0]):
    ...
```

````python
# enumerate() example
tasks = ["Write code", "Review PR", "Run tests", "Deploy"]

# 0-based index
print("--- 0-based ---")
for idx, task in enumerate(tasks):
    print(f"Task {idx}: {task}")

# 1-based index
print("\n--- 1-based ---")
for idx, task in enumerate(tasks, start=1):
    print(f"Step #{idx}: {task}")
````

### 9. `zip()`: Parallel Iteration

**Explanation**: Iterates over several iterables in parallel, pairing corresponding elements into tuples. Stops at the shortest iterable (or use itertools.zip_longest).

**Syntax**:
```python
for item1, item2 in zip(iter1, iter2):
    ...
```

````python
# zip() example
students = ["Alice", "Bob", "Charlie"]
scores = [95, 88, 92]
grades = ["A+", "B+", "A"]

for name, score, grade in zip(students, scores, grades):
    print(f"{name:<8} scored {score:>3} (Grade: {grade})")

# Constructing dictionaries with zip:
gradebook = dict(zip(students, scores))
print("\nGradebook dictionary:", gradebook)
````

### 10. Nested Loops & Multi-Level Breaks

**Explanation**: A loop inside another loop. To break out of outer loops from an inner loop, use a boolean flag, helper function with return, or loop else.

**Syntax**:
```python
# Method 1: Flag
# Method 2: Return from function (Cleanest)
```

````python
# Breaking out of nested loops cleanly using a function:
matrix = [
    [1, 2, 3],
    [4, 99, 6],
    [7, 8, 9]
]
target = 99

def find_in_matrix(mat, val):
    for r_idx, row in enumerate(mat):
        for c_idx, cell in enumerate(row):
            if cell == val:
                return (r_idx, c_idx)
    return None

coords = find_in_matrix(matrix, target)
print(f"Found {target} at row {coords[0]}, col {coords[1]}")
````
