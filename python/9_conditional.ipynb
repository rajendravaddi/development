## 1. Truth Value Testing (Truthiness vs Falsiness)

In Python, every object has an inherent boolean truth value. In an `if` condition, Python implicitly calls `bool(x)`:

### Falsy Objects in Python:
* Constants: `None`, `False`
* Zero of any numeric type: `0`, `0.0`, `0j`, `Decimal(0)`, `Fraction(0, 1)`
* Empty sequences and collections: `""`, `()`, `[]`, `{}`, `set()`, `range(0)`
* Custom objects that define `__bool__()` returning `False` or `__len__()` returning `0`.

**Everything else evaluates to `True` (Truthy).**

````python
# 1. Truthiness and Falsiness Examples
falsy_values = [None, False, 0, 0.0, "", (), [], {}, set()]

print("--- Testing Falsy Values ---")
for val in falsy_values:
    print(f"bool({val!r:<10}) -> {bool(val)}")
````

## 2. Standard `if`, `if-else`, and `if-elif-else`

Conditional structures allow executing code blocks conditionally based on expressions.

### The Simple `if` Statement

**Explanation**: Executes the indented block of code only if the condition evaluates to True. If False, the block is skipped.

**Syntax**:
```python
if condition:
    # code block executed when condition is True
```

````python
# Simple if example
temperature = 32

if temperature > 30:
    print(f"It is {temperature}°C - Stay hydrated!")
print("This line always runs.")
````

### The `if-else` Statement

**Explanation**: Provides an alternative branch: executes block 1 if the condition is True, or block 2 if the condition is False.

**Syntax**:
```python
if condition:
    # executed if True
else:
    # executed if False
```

````python
# if-else example
age = 17

if age >= 18:
    print("Eligible to vote.")
else:
    print(f"Not eligible to vote. Wait {18 - age} more year(s).")
````

### The `if-elif-else` Ladder

**Explanation**: Evaluates multiple conditions in sequential top-down order. The first branch whose condition evaluates to True executes, and all remaining branches are skipped. If none match, the optional else executes.

**Syntax**:
```python
if condition_1:
    # block 1
elif condition_2:
    # block 2
elif condition_3:
    # block 3
else:
    # fallback block
```

````python
# if-elif-else example
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score {score} yields grade: {grade}")
````

### Nested Conditional Statements

**Explanation**: An if, elif, or else statement placed inside another if, elif, or else block to evaluate complex hierarchical criteria.

**Syntax**:
```python
if outer_condition:
    if inner_condition:
        # inner block
    else:
        # inner alternative
else:
    # outer alternative
```

````python
# Nested conditionals example
has_ticket = True
has_id = False

if has_ticket:
    if has_id:
        print("Welcome inside the venue!")
    else:
        print("Ticket is valid, but government photo ID is required.")
else:
    print("No ticket found. Entry denied.")
````

### The Ternary Operator (Conditional Expression)

**Explanation**: A one-line expression that evaluates and returns one of two values based on a boolean condition.

**Syntax**:
```python
value_if_true if condition else value_if_false
```

````python
# Ternary operator example
num = 15
parity = "Even" if num % 2 == 0 else "Odd"
print(f"{num} is {parity}")
````

### Short-Circuiting in Compound Conditions

**Explanation**: In 'A and B', if A is False, B is never evaluated. In 'A or B', if A is True, B is never evaluated. Useful to guard against ZeroDivisionError or AttributeError.

**Syntax**:
```python
if denominator != 0 and (numerator / denominator) > 1:
    ...
```

````python
# Short-circuiting safety guards
numerator = 10
# denominator = 0
denominator = 3

# SAFE: denominator != 0 evaluates to False, so division is NEVER run!
if denominator != 0 and (numerator / denominator) > 2:
    print("Result is greater than 2")
else:
    print("Zero denominator safely avoided without crashing!")
````

## 3. Structural Pattern Matching: `match-case` (Python 3.10+)

Introduced in PEP 634, `match-case` is far more powerful than traditional `switch-case` statements found in other languages. It can match values, types, shapes, unpack sequences, inspect dictionaries, and match class attributes.

### Basic Literal Matching & Wildcard (`_`)

**Explanation**: Matches an expression against literal values. The underscore `_` acts as a wildcard (default/catch-all case).

**Syntax**:
```python
match expression:
    case literal_1:
        # code
    case literal_2:
        # code
    case _:
        # default case
```

````python
# status = 200
# status = 404
status = 418

# Literal match-case example
match status:
    case 200:
        print("OK: Request succeeded.")
    case 301:
        print("Moved Permanently: Redirection.")
    case 400:
        print("Bad Request: Client error.")
    case 404:
        print("Not Found: Resource does not exist.")
    case 500:
        print("Internal Server Error.")
    case _:
        print(f"Unknown status code: {status}")
````

### Multiple Patterns with Pipe (`|`) & Guard Clauses (`if`)

**Explanation**: Use pipe `|` to match any one of several patterns (OR). Add an `if` guard clause to check dynamic boolean conditions within a case.

**Syntax**:
```python
match value:
    case pattern_1 | pattern_2:
        # code
    case pattern if condition:
        # code
```

````python
# Combined patterns and guards example
def categorize_number(val: int):
    match val:
        case 0 | 1:
            return "Binary digit (0 or 1)"
        case n if n < 0:
            return f"Negative number: {n}"
        case n if n % 2 == 0:
            return f"Positive even number: {n}"
        case n:
            return f"Positive odd number: {n}"

print(categorize_number(1))
print(categorize_number(-42))
print(categorize_number(18))
print(categorize_number(23))
````

### Sequence Pattern Matching & Unpacking

**Explanation**: Destructures tuples, lists, or sequences into variables, matching structure and length with optional rest parameter `*rest`.

**Syntax**:
```python
match command:
    case [action]:
        ...
    case [action, target]:
        ...
    case [action, *targets]:
        ...
```

````python
# Sequence matching example
def process_command(cmd: list):
    match cmd:
        case ["quit"]:
            print("Exiting application...")
        case ["load", filename]:
            print(f"Loading file: '{filename}'")
        case ["move", x, y]:
            print(f"Moving to coordinates ({x}, {y})")
        case ["delete", *files]:
            print(f"Deleting files: {files}")
        case _:
            print(f"Invalid command format: {cmd}")

process_command(["quit"])
process_command(["load", "data.csv"])
process_command(["move", 100, 250])
process_command(["delete", "f1.txt", "f2.txt", "f3.txt"])
process_command(["unknown", 1, 2, 3])
````

### Mapping (Dictionary) Pattern Matching

**Explanation**: Inspects dictionary keys and values. Extra keys in the dictionary are ignored unless captured.

**Syntax**:
```python
match data:
    case {'type': 'user', 'name': name}:
        ...
    case {'type': 'admin'}:
        ...
```

````python
# Mapping match-case example
def handle_event(event: dict):
    match event:
        case {"type": "click", "x": x, "y": y}:
            print(f"Click registered at ({x}, {y})")
        case {"type": "keypress", "key": key}:
            print(f"Key pressed: '{key}'")
        case {"type": "error", "code": code, "message": msg}:
            print(f"Error {code}: {msg}")
        case _:
            print("Unrecognized event structure.")

handle_event({"type": "click", "x": 10, "y": 20, "timestamp": 1234567})
handle_event({"type": "keypress", "key": "Enter"})
handle_event({"type": "error", "code": 403, "message": "Forbidden"})
````

### Class / Object Pattern Matching

**Explanation**: Matches instances of specific classes and extracts attributes directly inside case patterns.

**Syntax**:
```python
match obj:
    case Point(x=0, y=0):
        ...
    case Point(x=x, y=y):
        ...
```

````python
# Class pattern matching example
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

def inspect_point(p):
    match p:
        case Point(x=0, y=0):
            print("Point is at the Origin (0, 0)")
        case Point(x=0, y=y):
            print(f"Point is on the Y-axis at y={y}")
        case Point(x=x, y=0):
            print(f"Point is on the X-axis at x={x}")
        case Point(x=x, y=y) if x == y:
            print(f"Point is on diagonal line y=x at ({x}, {y})")
        case Point(x=x, y=y):
            print(f"General point at ({x}, {y})")
        case _:
            print("Not a Point object")

inspect_point(Point(0, 0))
inspect_point(Point(0, 5))
inspect_point(Point(4, 4))
inspect_point(Point(3, 7))
````
