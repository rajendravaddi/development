## 1. Definition and Creation

A **string** in Python is an **immutable sequence of Unicode characters** (represented by the `str` type). Python strings can represent text in any language, emojis, symbols, and whitespace.

### How to Create Strings:
* **Single quotes**: `'Hello'`
* **Double quotes**: `"Hello"` (equivalent to single quotes; useful if string contains single quotes)
* **Triple quotes**: `'''...'''` or `"""..."""` (multi-line strings and docstrings)
* **Constructor**: `str(...)` converts objects into their string representation.

````python
# 1. String Creation Examples
s_single = 'Hello, Python!'
s_double = "Data Science"
s_multiline = """This is a multi-line string.
It preserves line breaks and
    indentation automatically."""

print("Single quote string:", s_single)
print("Double quote string:", s_double)
print("Multi-line string:\n", s_multiline)

````

````python
# Using str() constructor
s_num = str(12345)
s_bool = str(True)
s_list = str([1, 2, 3])

print("Converted from int:", s_num, type(s_num))
print("Converted from bool:", s_bool, type(s_bool))
print("Converted from list:", s_list, type(s_list))
````

## 2. How Memory is Managed (String Interning & PEP 393)

### How Python Manages String Memory:
1. **Unicode Representation (PEP 393 - Flexible String Representation)**:
   Python dynamically chooses the smallest byte-size per character depending on the contents:
   * **Latin-1** (1 byte / char): for ASCII & Latin-1 characters (up to `\u00FF`).
   * **UCS-2** (2 bytes / char): for characters requiring up to 16 bits (up to `\uFFFF`).
   * **UCS-4** (4 bytes / char): for emoji and rare scripts (up to `\U0010FFFF`).
   This ensures memory efficiency.

2. **String Interning**:
   Python caches / re-uses certain string instances in memory so they share the exact same `id()` (memory address).
   * Identifiers (variable names, method names, dictionary keys) and small compile-time ASCII strings without whitespace are automatically interned.
   * You can explicitly intern strings using `sys.intern()`. Interning speeds up comparisons from $O(n)$ character checks to $O(1)$ pointer comparison using `is`.

````python
# Automatic interning of compile-time ASCII constants
str1 = "python_language"
str2 = "python_language"
print("str1 is str2:", str1 is str2)               # True (interned automatically)
print("id(str1) == id(str2):", id(str1) == id(str2))
````

````python

# Strings created at runtime or with spaces may not be automatically interned
s_dynamic_1 = "".join(["hello", " world"])
s_dynamic_2 = "".join(["hello", " world"])
print("\nBefore sys.intern:")
print("s_dynamic_1 == s_dynamic_2:", s_dynamic_1 == s_dynamic_2) # True (values match)
print("s_dynamic_1 is s_dynamic_2:", s_dynamic_1 is s_dynamic_2) # False (different objects)
````

````python
import sys

# Explicit interning via sys.intern()
s_interned_1 = sys.intern(s_dynamic_1)
s_interned_2 = sys.intern(s_dynamic_2)
print("\nAfter sys.intern:")
print("s_interned_1 is s_interned_2:", s_interned_1 is s_interned_2) # True (same address)
````

````python

import sys
# Memory footprint across encodings (PEP 393)
ascii_str = "hello"
unicode_str = "привет"  # Cyrillic
emoji_str = "👋🌍"      # Emojis

print("\nMemory sizes (sys.getsizeof):")
print(f"ASCII '{ascii_str}': {sys.getsizeof(ascii_str)} bytes")
print(f"Cyrillic '{unicode_str}': {sys.getsizeof(unicode_str)} bytes")
print(f"Emoji '{emoji_str}': {sys.getsizeof(emoji_str)} bytes")
````

## 3. String Immutability

In Python, **strings are strictly immutable**. Once a string object is created in memory, its characters **cannot be modified, added, or deleted in place**.

### Why are strings immutable?
* **Security & Thread Safety**: Can be passed across functions and threads without fear of mutation.
* **Hashability**: Because strings are immutable, their hash value never changes. This enables strings to be used as dictionary keys and set elements.
* **Optimization**: Allows Python to safely intern and cache strings without side effects.

> Any operation that looks like it modifies a string (e.g., `.replace()`, `.upper()`, `+=`) actually creates a **brand-new string object** in memory.

````python
text = "Python"

# Attempting to modify an in-place character raises TypeError
try:
    text[0] = "J"
except TypeError as e:
    print(f"Error caught: {e}")
````

````python

text = "Python"

# Creating a modified version requires constructing a new string
old_id = id(text)
text = "J" + text[1:]
new_id = id(text)

print(f"Modified text: {text}")
print(f"Old object ID: {old_id} vs New object ID: {new_id} (IDs differ!)")
````

````python


# Strings can be dictionary keys because they are immutable (hashable)
cache = {"username": "alice123", "token": "xyz987"}
print(f"String hash: {hash('username')}")
````

## 4. String Indexing and Slicing

Strings are ordered sequences, meaning each character has an index.

### 1. Indexing (0-based)
* **Positive index**: Starts from `0` on the left.
* **Negative index**: Starts from `-1` on the right (last character).

```text
 String:   P   y   t   h   o   n
 Index:    0   1   2   3   4   5
-Index:   -6  -5  -4  -3  -2  -1
```


````python
word = "PythonProgramming"

print("--- Indexing ---")
print(f"First character (word[0])   : {word[0]}")
print(f"Sixth character (word[5])   : {word[5]}")
print(f"Last character (word[-1])   : {word[-1]}")
print(f"Second to last (word[-2])   : {word[-2]}")
````


### 2. Slicing Syntax: `string[start:stop:step]`
* `start`: Inclusive starting index (default is `0` or end if step < 0).
* `stop`: Exclusive ending index (slice stops right before this index).
* `step`: Amount by which the index increases (default is `1`). Negative step reverses direction.

````python
word = "PythonProgramming"

print("\n--- Slicing: [start:stop:step] ---")
print(f"word[0:6]     : {word[0:6]}")       # 'Python' (indices 0 to 5)
print(f"word[:6]      : {word[:6]}")        # 'Python' (defaults start to 0)
print(f"word[6:]      : {word[6:]}")        # 'Programming' (from 6 to end)
print(f"word[::2]     : {word[::2]}")       # Every 2nd character: 'PtoPormn'
print(f"word[1::2]    : {word[1::2]}")      # Odd indices: 'yhnrgamg'

````

````python
word = "PythonProgramming"

print("\n--- Reversal with Negative Step ---")
print(f"word[::-1]    : {word[::-1]}")      # Complete string reversed!
print(f"word[5::-1]   : {word[5::-1]}")     # Reverse first 6 characters: 'nohtyP'
````

## 5. Concatenation and Repetition

* **Concatenation (`+`)**: Combines two or more strings together.
* **Implicit concatenation**: String literals written next to each other are joined automatically at compile time.
* **Repetition (`*`)**: Repeats a string a specified number of times.

> **Performance Tip**: Repeatedly appending strings in a loop with `+=` creates $O(n^2)$ memory copies. Always collect substrings in a list and join them with `''.join(list)` for $O(n)$ performance.

````python
str1 = "Data"
str2 = "Science"

# 1. Concatenation with '+'
full_str = str1 + " " + str2
print(f"Concatenation: {full_str}")
````

````python
# 2. Implicit string literal concatenation (compile-time)
intro = ("Welcome to Python! "
         "This long sentence is split across lines "
         "without any + operator.")
print(f"Implicit:\n{intro}")
````

````python
# 3. Repetition with '*'
separator = "=" * 30
header = "REPORT"
print(f"\n{separator}\n{header.center(30)}\n{separator}")

````

````python
# 4. Best Practice for loop concatenation: str.join()
# BAD: string += item (allocates new string on each iteration)
# GOOD: join()
parts = [f"item_{i}" for i in range(5)]
efficient_joined = ", ".join(parts)
print(f"\nEfficiently joined: {efficient_joined}")
````

## 6. String Formatting

Python has evolved through multiple ways of formatting strings:

1. **Old-style `%` (printf-style)**: Inherited from C (e.g., `"%s is %d" % (name, age)`).
2. **`str.format()` (Python 2.6+)**: Positional & keyword arguments formatting.
3. **Formatted String Literals / f-strings (Python 3.6+)**: The modern, fastest, and most readable approach.
4. **`string.Template`**: Safer alternative for user-supplied templates (avoids code execution risks).

````python
name = "Alice"
score = 92.4567
rank = 1

# 1. Old-Style % Formatting
fmt_percent = "Hello %s, score: %.2f, rank: %d" % (name, score, rank)
print("1. % operator :", fmt_percent)
````

````python
# 2. str.format() method
fmt_format = "Hello {}, score: {:.2f}, rank: {}".format(name, score, rank)
fmt_named  = "Hello {n}, score: {s:.2f}".format(n=name, s=score)
print("2. str.format :", fmt_format)
````

````python
# 3. Formatted string literals (f-strings) - Recommended!
fmt_fstring = f"Hello {name.upper()}, score: {score:.2f}, rank: {rank}"
print("3. f-string   :", fmt_fstring)
````

````python
# Advanced f-string formatting specifiers
pi = 3.1415926535
number = 1000000000
percentage = 0.856

print(f"\n--- Format Specifiers ---")
print(f"Decimals    : {pi:.3f}")           # 3 decimal places
print(f"Grouping    : {number:,}")          # 1,000,000,000
print(f"Percentage  : {percentage:.1%}")    # 85.6%
print(f"Padding/Align: |{name:<10}| |{name:^10}| |{name:>10}|") # Left, Center, Right
print(f"Hex & Binary: {255:#x}, {255:#b}") # 0xff, 0b11111111
````

````python
# 4. string.Template (Safe for untrusted user inputs)
from string import Template
tmpl = Template("Hello, $user! You have $msgs unread messages.")
safe_message = tmpl.substitute(user=name, msgs=5)
print(f"\n4. Template   : {safe_message}")
````

## 7. Escape Characters, Raw Strings, and Byte Strings

### Escape Sequences
An escape character begins with a backslash `\` to represent characters that are otherwise difficult or impossible to type:

| Escape Sequence | Meaning |
| :--- | :--- |
| `\n` | Newline (line feed) |
| `\t` | Horizontal Tab |
| `\\` | Backslash literal |
| `\'` | Single quote |
| `\"` | Double quote |
| `\r` | Carriage return |
| `\b` | Backspace |
| `\N{name}` | Unicode character by name (e.g. `\N{GREEK CAPITAL LETTER DELTA}`) |
| `\uXXXX` | 16-bit hex Unicode (e.g. `\u0394`) |
| `\UXXXXXXXX` | 32-bit hex Unicode (e.g. `\U0001F600`) |

### Raw Strings (`r"..."`)
Prefixing a string with `r` or `R` suppresses escape interpretation. Essential for **regular expressions** and **Windows file paths**.

````python
# Standard escape sequences
escaped = "First Line\nSecond Line\t(Tabbed)\nHe said: \"Python is awesome!\""
print("--- Escaped String ---")
print(escaped)
````

````python
# Unicode escape sequences
print("\n--- Unicode Escapes ---")
print("Named unicode : \N{FIRE} \N{ROCKET}")
print("16-bit hex    : \u03B1 + \u03B2 = \u03B3")  # α + β = γ
print("32-bit hex    : \U0001F600")                # 😀
````

````python
# Raw string vs regular string with backslashes
path_regular = "C:\\Users\\Admin\\Documents"
path_raw = r"C:\Users\Admin\Documents"
print("\n--- Raw Strings ---")
print(f"Regular escaped path : {path_regular}")
print(f"Raw string path      : {path_raw}")
print(f"Are they equal?      : {path_regular == path_raw}")
````

## 8. All Built-in String Methods

Every built-in Python string method is detailed below with:
* **Functionality sentence**: A clear explanation of what the method does.
* **Syntax**: Parameters, defaults, and return type.
* **Code cell**: Executable code demonstrating its usage and variations.

### `upper()`

**Functionality**: Converts all lowercase characters in the string into uppercase and returns the new string.

**Syntax**:
```python
str.upper() -> str
```

````python
# upper() example
text = "hello, python 123!"
upper_text = text.upper()

print("Original:", text)
print("upper() :", upper_text)
````

### `lower()`

**Functionality**: Converts all uppercase characters in the string into lowercase and returns the new string.

**Syntax**:
```python
str.lower() -> str
```

````python
# lower() example
text = "HELLO, PYTHON 123!"
lower_text = text.lower()

print("Original:", text)
print("lower() :", lower_text)
````

### `capitalize()`

**Functionality**: Converts the very first character of the string to uppercase and converts all other characters to lowercase.

**Syntax**:
```python
str.capitalize() -> str
```

````python
# capitalize() example
s1 = "python PROGRAMMING"
s2 = "123 hello"

print("s1.capitalize():", s1.capitalize())  # Only 'P' is capitalized
print("s2.capitalize():", s2.capitalize())  # First char is digit, rest lowercase
````

### `title()`

**Functionality**: Converts the first character of each word to uppercase and the remaining letters to lowercase (Title Case).

**Syntax**:
```python
str.title() -> str
```

````python
# title() example
heading = "welcome to the complete python guide"
print("title():", heading.title())

# Note on apostrophes with title():
print("they're.title():", "they're".title())  # 'They'Re' (apostrophe treated as word boundary)
````

### `swapcase()`

**Functionality**: Inverts the case of all cased characters: converts uppercase letters to lowercase and lowercase to uppercase.

**Syntax**:
```python
str.swapcase() -> str
```

````python
# swapcase() example
text = "PyThOn Is CoOl"
swapped = text.swapcase()

print("Original:", text)
print("swapcase():", swapped)
````

### `casefold()`

**Functionality**: Returns a casefolded string for aggressive, caseless string matching. More powerful than lower() as it handles international Unicode characters.

**Syntax**:
```python
str.casefold() -> str
```

````python
# casefold() example
# German sharp s: 'ß' lower() is 'ß', but casefold() turns it into 'ss'
german_word = "Straße"

print("lower()    :", german_word.lower())
print("casefold() :", german_word.casefold())

# Caseless comparison
print("Matching with 'strasse':", german_word.casefold() == "strasse")
````

### `count()`

**Functionality**: Returns the number of non-overlapping occurrences of a substring within the optional slice [start:end].

**Syntax**:
```python
str.count(sub[, start[, end]]) -> int
```

````python
# count() example
message = "banana band and bandana"

print("Count of 'an'       :", message.count("an"))
print("Count of 'an' from 5:", message.count("an", 5))       # start at index 5
print("Count of 'an' (5:15):", message.count("an", 5, 15))   # start at 5, stop before 15
````

### `find()`

**Functionality**: Returns the lowest index where the substring is found within [start:end]. Returns -1 if substring is NOT found.

**Syntax**:
```python
str.find(sub[, start[, end]]) -> int
```

````python
# find() example
phrase = "programming in python"

print("find('in')    :", phrase.find("in"))      # 8 (first occurrence in 'programming')
print("find('in', 10):", phrase.find("in", 10))  # 12 (the word 'in')
print("find('java')  :", phrase.find("java"))    # -1 (not found)
````

### `rfind()`

**Functionality**: Returns the highest (rightmost) index where the substring is found within [start:end]. Returns -1 if NOT found.

**Syntax**:
```python
str.rfind(sub[, start[, end]]) -> int
```

````python
# rfind() example
path = "/home/user/documents/report.pdf"

last_slash = path.rfind("/")
print("rfind('/') index:", last_slash)
print("Filename extracted:", path[last_slash + 1:])
print("rfind('xyz')     :", path.rfind("xyz"))  # -1
````

### `index()`

**Functionality**: Like find(), but raises a ValueError exception if the substring is not found instead of returning -1.

**Syntax**:
```python
str.index(sub[, start[, end]]) -> int
```

````python
# index() example
text = "learning python"

print("index('python'):", text.index("python"))  # 9

# Raises ValueError if not found
try:
    text.index("rust")
except ValueError as err:
    print("Error caught:", err)
````

### `rindex()`

**Functionality**: Like rfind(), but raises a ValueError exception if the substring is not found instead of returning -1.

**Syntax**:
```python
str.rindex(sub[, start[, end]]) -> int
```

````python
# rindex() example
text = "apple pie, apple tart"

print("rindex('apple'):", text.rindex("apple"))  # 11 (rightmost)

try:
    text.rindex("banana")
except ValueError as err:
    print("Error caught:", err)
````

### `startswith()`

**Functionality**: Returns True if the string starts with the specified prefix (or any prefix in a tuple of prefixes); otherwise False.

**Syntax**:
```python
str.startswith(prefix[, start[, end]]) -> bool
```

````python
# startswith() example
url = "https://www.python.org"

print("startswith('https://') :", url.startswith("https://"))
print("startswith('http://')  :", url.startswith("http://"))

# Passing a tuple of prefixes
valid_schemes = ("http://", "https://", "ftp://")
print("startswith(tuple)      :", url.startswith(valid_schemes))

# With start slice
print("startswith('www', 8)   :", url.startswith("www", 8))
````

### `endswith()`

**Functionality**: Returns True if the string ends with the specified suffix (or any suffix in a tuple of suffixes); otherwise False.

**Syntax**:
```python
str.endswith(suffix[, start[, end]]) -> bool
```

````python
# endswith() example
filename = "data_analysis.csv"

print("endswith('.csv') :", filename.endswith(".csv"))
print("endswith('.xlsx'):", filename.endswith(".xlsx"))

# Checking multiple extensions with a tuple
image = "photo.png"
valid_extensions = (".jpg", ".jpeg", ".png", ".gif")
print("Is valid image? :", image.endswith(valid_extensions))
````

### `isalpha()`

**Functionality**: Returns True if all characters in the string are alphabetic (letters) and the string is not empty; otherwise False.

**Syntax**:
```python
str.isalpha() -> bool
```

````python
# isalpha() example
print("'Python'.isalpha()   :", "Python".isalpha())     # True
print("'Python3'.isalpha()  :", "Python3".isalpha())    # False (contains digit)
print("'Hello World'.isalpha():", "Hello World".isalpha())# False (contains space)
print("''.isalpha()         :", "".isalpha())           # False (empty string)
````

### `isdigit()`

**Functionality**: Returns True if all characters in the string are digits (including superscripts like ² and subscripts) and the string is not empty.

**Syntax**:
```python
str.isdigit() -> bool
```

````python
# isdigit() example
print("'12345'.isdigit() :", "12345".isdigit())   # True
print("'12.34'.isdigit() :", "12.34".isdigit())   # False (decimal point is not a digit)
print("'2²'.isdigit()    :", "2²".isdigit())      # True (superscript 2 is a digit!)
print("'-10'.isdigit()   :", "-10".isdigit())     # False (minus sign is not a digit)
````

### `isalnum()`

**Functionality**: Returns True if all characters in the string are alphanumeric (either letters or digits) and not empty; otherwise False.

**Syntax**:
```python
str.isalnum() -> bool
```

````python
# isalnum() example
print("'Python3'.isalnum()    :", "Python3".isalnum())      # True
print("'User_123'.isalnum()   :", "User_123".isalnum())     # False (underscore is not alphanumeric)
print("'abc'.isalnum()        :", "abc".isalnum())          # True
print("'999'.isalnum()        :", "999".isalnum())          # True
````

### `isdecimal()`

**Functionality**: Returns True if all characters in the string are decimal characters (strictly base-10 digits 0-9) and not empty.

**Syntax**:
```python
str.isdecimal() -> bool
```

````python
# isdecimal() example
print("'12345'.isdecimal() :", "12345".isdecimal())  # True
print("'2²'.isdecimal()    :", "2²".isdecimal())     # False (superscript 2 is digit, NOT decimal!)
print("'½'.isdecimal()     :", "½".isdecimal())      # False
````

### `isnumeric()`

**Functionality**: Returns True if all characters in the string are numeric characters (includes base-10 digits, fractions like ½, roman numerals, superscript/subscript).

**Syntax**:
```python
str.isnumeric() -> bool
```

````python
# isnumeric() example
print("'12345'.isnumeric() :", "12345".isnumeric())  # True
print("'2²'.isnumeric()    :", "2²".isnumeric())     # True
print("'½'.isnumeric()     :", "½".isnumeric())      # True (fraction is numeric!)
print("'IV'.isnumeric()    :", "IV".isnumeric())     # False (standard ASCII letters)
print("'Ⅳ'.isnumeric()    :", "Ⅳ".isnumeric())     # True (Unicode Roman Numeral char Ⅳ)
````

### `isspace()`

**Functionality**: Returns True if there are only whitespace characters (spaces, tabs, newlines, carriage returns) in the string and not empty.

**Syntax**:
```python
str.isspace() -> bool
```

````python
# isspace() example
print("'   '.isspace()       :", "   ".isspace())       # True

print("'  a  '.isspace()     :", "  a  ".isspace())     # False (contains 'a')
print("''.isspace()          :", "".isspace())          # False
````

### `islower()`

```python
str.islower() -> bool
```

````python
# islower() example
print("'hello'.islower()     :", "hello".islower())      # True
print("'hello world! 123'.islower() :", "hello world! 123".islower()) # True (ignores digits & punctuation)
print("'Hello'.islower()     :", "Hello".islower())      # False
print("'123!'.islower()      :", "123!".islower())       # False (no cased characters)
````

```python
str.isupper() -> bool
```

### `istitle()`

### `isidentifier()`

**Functionality**: Returns True if the string is a syntactically valid identifier according to Python language rules (e.g., variable or function name).

**Syntax**:
```python
str.isidentifier() -> bool
```

````python
# isidentifier() example
print("'my_var'.isidentifier()   :", "my_var".isidentifier())    # True
print("'_private'.isidentifier() :", "_private".isidentifier())  # True
print("'2nd_var'.isidentifier()  :", "2nd_var".isidentifier())   # False (starts with digit)
print("'my-var'.isidentifier()   :", "my-var".isidentifier())    # False (hyphen not allowed)
print("'for'.isidentifier()      :", "for".isidentifier())       # True (valid identifier syntax, though a reserved keyword)
````

````python
# rstrip() example
text = "Welcome to Python!    \n\t"
print("rstrip() whitespace :", repr(text.rstrip()))

filename = "document.txt.bak"
print("rstrip('.bak')      :", filename.rstrip(".bak"))
````

### `split()`

### `partition()`

**Functionality**: Searches for the first occurrence of sep and returns a 3-tuple containing: (part before sep, sep, part after sep). If sep is not found, returns (string, '', '').

**Syntax**:
```python
str.partition(sep) -> tuple[str, str, str]
```

````python
# partition() example
email = "user.name@example.com"
user, sep, domain = email.partition("@")

print(f"User: '{user}', Sep: '{sep}', Domain: '{domain}'")

# If separator is not found
not_found = "plain_text".partition(":")
print("Not found partition:", not_found)
````


````python
# rpartition() example
url = "https://www.example.com/blog/python/intro"
base, sep, slug = url.rpartition("/")

print(f"Base: '{base}'")
print(f"Sep : '{sep}'")
print(f"Slug: '{slug}'")
````

### `zfill()`

**Functionality**: Pads a numeric string with zeros '0' on the left until it reaches the specified width. Correctly handles leading '+' and '-' signs.


````python
# zfill() example
print("'42'.zfill(5)    :", "42".zfill(5))      # '00042'
print("'-42'.zfill(5)   :", "-42".zfill(5))     # '-0042' (preserves sign position)
print("'+42'.zfill(5)   :", "+42".zfill(5))     # '+0042'
print("'12345'.zfill(3) :", "12345".zfill(3))   # '12345' (no truncation if width <= len)
````

### `translate()`

**Functionality**: Returns a copy of the string where each character has been mapped through the given translation table (typically created by str.maketrans()).

**Syntax**:
```python

```

````python
# translate() example
# 1. Leetspeak mapping
trans_table = str.maketrans("aeiot", "43107")
original = "the quick brown fox jumps over the lazy dog"
print("Leet translated:", original.translate(trans_table))

# 2. Removing punctuation efficiently
import string
remove_punct = str.maketrans("", "", string.punctuation)
cleaned = "Hello, World! How are you doing today?".translate(remove_punct)
print("Cleaned text   :", cleaned)
````

### `encode()`

**Functionality**: Encodes the string into a bytes object using the specified encoding (default 'utf-8'). Handles encoding errors via the errors parameter.

**Syntax**:
```python

```

````python
# encode() example
text = "Python 🐍 & Café"

# UTF-8 encoding
encoded_utf8 = text.encode("utf-8")
print("UTF-8 bytes:", encoded_utf8)

# ASCII encoding with error handlers: 'ignore', 'replace', 'backslashreplace'
ascii_ignore = text.encode("ascii", errors="ignore")
ascii_replace = text.encode("ascii", errors="replace")
print("ASCII ignore :", ascii_ignore)
print("ASCII replace:", ascii_replace)
````

## 9. Modern String Features & Additions

Python continues to introduce powerful string features across recent releases:

### 1. Self-Documenting f-strings with `=` (Python 3.8+)
Writing `{var=}` prints both the expression name and its evaluated value, perfect for rapid debugging.

### 2. PEP 701 - Syntactic Formalization of f-strings (Python 3.12+)
Prior to Python 3.12, f-strings could not contain backslashes inside expressions (e.g. `f"{ '\n'.join(items) }"` caused a SyntaxError) and quotes could not be reused without switching between single and double quotes. In Python 3.12+, all quote nesting and backslashes inside expressions are fully valid!

### 3. String Encoding & Decoding (`encode()` / `decode()`)
Strings (`str`) represent human text (Unicode), while `bytes` represent raw binary data. Converting between them requires explicit encoding/decoding (`utf-8`).

````python
# 1. Self-documenting f-strings (Python 3.8+)
x = 42
y = 100
calc = x * y
print("--- Self-documenting f-strings ({expr=}) ---")
print(f"{x=}, {y=}, {calc=}")
print(f"{x * 2 + y / 10=}")
````

````python
# 2. String Encoding & Decoding
print("\n--- Encoding (str -> bytes) & Decoding (bytes -> str) ---")
original_text = "Python 🐍 is awesome!"
# Encode str to bytes (UTF-8)
encoded_bytes = original_text.encode("utf-8")
print(f"Encoded bytes : {encoded_bytes}")
print(f"Type of bytes : {type(encoded_bytes)}")
````

````python

# Decode bytes back to str
decoded_text = encoded_bytes.decode("utf-8")
print(f"Decoded text  : {decoded_text}")
print(f"Matches?      : {original_text == decoded_text}")
````
