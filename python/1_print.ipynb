# print()

Syntax:
```python
print(*objects, sep=' ', end='\n', file=None, flush=False)
```
Parameter|Description|Default value
---|---|---
*objects|One or more values to display|No objects
sep|Separator between multiple values|' ' (space)
end|What to append after printing|'\n' (newline)
file|Where to send the output|None (usually the screen)
flush|Whether to force the output to be written immediately|False

````python
print("Hello")
````

## objects — Values to print
You can pass one or more objects to the print() function.

````python
a = 10
b = "Hello"
print(a,b)
````

## sep — Separator between values

The sep parameter specifies the string inserted between multiple objects.

sep controls what appears between values. It does not control what appears at the end.

````python
print("2026", "10", "28", sep="-")
````

## end — Ending character or string

By default, print() adds a newline (\n) after printing. 

This means the next print() call starts on a new line.

````python
print("Hello")
print("Python")
````

````python
print("Hello", end=" ")
print("Python")
````

## file — Destination of the output
The file parameter specifies where the output should be written. 

By default, output goes to standard output, usually your terminal or console.

````python
# Here the output goes into the file
with open("1_print_file.txt", "w") as f:
    print("Hello, Python!", file=f)
````

## flush — Force output immediately

Normally, Python may buffer output before displaying or writing it. 

Setting flush=True requests that the output stream be flushed immediately.

````python
import time

print("Processing...", end=" ", flush=True)
time.sleep(2)
print("Done!")
````
