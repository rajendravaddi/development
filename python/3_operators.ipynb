## 1. Arithmetic Operators

Arithmetic operators perform standard mathematical operations on numeric types (`int`, `float`, `complex`).

| Operator | Name | Description | Example |
| :--- | :--- | :--- | :--- |
| `+` | Addition | Adds two operands (also concatenates sequences) | `a + b` |
| `-` | Subtraction | Subtracts the right operand from the left | `a - b` |
| `*` | Multiplication | Multiplies two operands (also repeats sequences) | `a * b` |
| `/` | True Division | Divides left by right; **always returns float** | `a / b` |
| `//` | Floor Division | Divides and truncates toward negative infinity | `a // b` |
| `%` | Modulus | Returns remainder of division | `a % b` |
| `**` | Exponentiation | Left operand raised to power of right operand | `a ** b` |
| `+x`, `-x` | Unary Plus / Minus | Retains or negates the sign of operand | `+a`, `-a` |

```python
# 1. Arithmetic Operators Example
a = 15
b = 4

print("--- Basic Arithmetic ---")
print(f"{a} + {b}  = {a + b}")     # Addition: 19
print(f"{a} - {b}  = {a - b}")     # Subtraction: 11
print(f"{a} * {b}  = {a * b}")     # Multiplication: 60
print(f"{a} / {b}  = {a / b}")     # True Division: 3.75 (always float)
print(f"{a} // {b} = {a // b}")    # Floor Division: 3
print(f"{a} % {b}  = {a % b}")     # Modulus: 3
print(f"{a} ** {b} = {a ** b}")    # Exponentiation: 15^4 = 50625
```

```python

# Unary operators
x = 10
print(f"\nUnary minus of {x}: {-x}")
print(f"Unary minus of {-x}: {-(-x)}")
```

```python

# Floor division and negative numbers (rounds down toward -infinity)
print(f"-15 // 4 = {-15 // 4}")     # -4 (not -3!)
print(f"-15 % 4  = {-15 % 4}")      # 1 (because -15 = 4 * (-4) + 1)
```

## 2. Comparison (Relational) Operators

Comparison operators compare values and always evaluate to a boolean (`True` or `False`).

| Operator | Meaning | Example |
| :--- | :--- | :--- |
| `==` | Equal to | `a == b` |
| `!=` | Not equal to | `a != b` |
| `>` | Greater than | `a > b` |
| `<` | Less than | `a < b` |
| `>=` | Greater than or equal to | `a >= b` |
| `<=` | Less than or equal to | `a <= b` |

### Chained Comparisons
Python supports chaining comparisons naturally, such as `a < b < c`, which is equivalent to `(a < b) and (b < c)`, but `b` is evaluated only once.

```python
# 2. Comparison Operators Example
x = 10
y = 20

print("--- Comparison Operators ---")
print(f"{x} == {y} : {x == y}")   # False
print(f"{x} != {y} : {x != y}")   # True
print(f"{x} > {y}  : {x > y}")    # False
print(f"{x} < {y}  : {x < y}")    # True
print(f"{x} >= {y} : {x >= y}")   # False
print(f"{x} <= {y} : {x <= y}")   # True
```

```python

print("\n--- Chained Comparisons ---")
score = 85
# Check if score is between 80 and 90
print(f"80 <= {score} < 90 : {80 <= score < 90}") # True
```

```python

# Equivalent to (score >= 80 and score < 90)
a, b, c = 5, 10, 15
print(f"{a} < {b} < {c}    : {a < b < c}")         # True
print(f"{a} < {b} > {c}    : {a < b > c}")         # False
```

## 3. Assignment Operators

Assignment operators bind values to variables or mutate objects in place.

* **Simple assignment**: `=`
* **Compound / Augmented assignment**: Applies an operation and re-assigns the result in one step (e.g., `x += y` is equivalent to `x = x + y` for immutables).

| Operator | Equivalent Expression |
| :--- | :--- |
| `+=` | `x = x + y` |
| `-=` | `x = x - y` |
| `*=` | `x = x * y` |
| `/=` | `x = x / y` |
| `//=` | `x = x // y` |
| `%=` | `x = x % y` |
| `**=` | `x = x ** y` |
| `&=`, `|=`, `^=` | Bitwise AND, OR, XOR assignment |
| `>>=`, `<<=` | Bitwise Shift assignment |

> **Note on Mutability**: For mutable objects like `list`, `+=` extends the list in-place via `extend()`, mutating the original instance.

```python
# 3. Assignment Operators Example
val = 10
print(f"Initial value: {val}")

val += 5   # val = val + 5 -> 15
print(f"After val += 5  : {val}")

val -= 3   # val = val - 3 -> 12
print(f"After val -= 3  : {val}")

val *= 2   # val = val * 2 -> 24
print(f"After val *= 2  : {val}")

val /= 4   # val = val / 4 -> 6.0
print(f"After val /= 4  : {val}")

val //= 2  # val = val // 2 -> 3.0
print(f"After val //= 2 : {val}")

val %= 2   # val = val % 2 -> 1.0
print(f"After val %= 2  : {val}")

val **= 4  # val = val ** 4 -> 1.0
print(f"After val **= 4 : {val}")
```

```python

# Mutability nuance with +=
print("\n--- In-place mutation with lists ---")
list1 = [1, 2]
list2 = list1
list1 += [3]
print(f"list1: {list1}")
print(f"list2: {list2} (list1 and list2 point to the same list object!)")
```

## 4. Logical (Boolean) Operators

Used to combine or invert boolean values and truthy/falsy expressions.

| Operator | Description |
| :--- | :--- |
| `and` | Returns `True` if **both** operands are truthy |
| `or` | Returns `True` if **at least one** operand is truthy |
| `not` | Inverts the truth value of the operand |

### Key Concepts:
1. **Falsy values in Python**: `0`, `0.0`, `""`, `[]`, `()`, `{}`, `set()`, `None`, and `False`. Everything else is truthy.
2. **Short-circuit Evaluation**:
   * For `A and B`: If `A` is falsy, Python returns `A` immediately without evaluating `B`.
   * For `A or B`: If `A` is truthy, Python returns `A` immediately without evaluating `B`.
3. **Return Values**: `and` and `or` return the **actual operand value** that decided the outcome, not just a bare boolean!

```python
# 4. Logical Operators Example
a = True
b = False

print("--- Logical Operators ---")
print(f"{a} and {b} : {a and b}")  # False
print(f"{a} or {b}  : {a or b}")   # True
print(f"not {a}     : {not a}")     # False
print(f"not {b}     : {not b}")     # True
```

```python

# 'or' returns the first truthy value or the last falsy value
user_input = ""
username = user_input or "Guest"
print(f"Default fallback: username = {username!r}")
```

```python

# 'and' returns the first falsy value or the last truthy value
result1 = "Hello" and [1, 2, 3]
result2 = "" and [1, 2, 3]
print(f"'Hello' and [1, 2, 3] : {result1}")  # [1, 2, 3]
print(f"'' and [1, 2, 3]      : {result2!r}")  # ''
```

## 5. Identity Operators

Identity operators check whether two variables point to the exact same object in computer memory (i.e., whether `id(a) == id(b)`).

| Operator | Description | Example |
| :--- | :--- | :--- |
| `is` | Evaluates to `True` if both variables refer to the identical object in memory | `a is b` |
| `is not` | Evaluates to `True` if both variables refer to different objects in memory | `a is not b` |

### `==` vs `is`:
* `==` tests **equality of content** (Do they have the same value?).
* `is` tests **identity of reference** (Are they literally the same object in RAM?).
* **Best practice**: Always use `is` / `is not` when comparing against singleton `None` (e.g., `x is None`).

```python
list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a

print("--- Content Equality (==) vs Reference Identity (is) ---")
print(f"list_a == list_b : {list_a == list_b}")   # True (same values)
print(f"list_a is list_b : {list_a is list_b}")   # False (different memory locations)
print(f"list_a is list_c : {list_a is list_c}")   # True (same object)
print(f"list_a is not list_b : {list_a is not list_b}") # True
```

```python

print(f"\nid(list_a) = {id(list_a)}")
print(f"id(list_b) = {id(list_b)}")
print(f"id(list_c) = {id(list_c)}")
```

```python

print("\n--- Comparing against None ---")
data = None
print(f"data is None     : {data is None}")
print(f"data is not None : {data is not None}")
```

## 6. Membership Operators

Membership operators test whether a specific element or sequence exists within an iterable (such as a string, list, tuple, set, or dictionary).

| Operator | Description | Example |
| :--- | :--- | :--- |
| `in` | Returns `True` if value is present in the collection | `x in seq` |
| `not in` | Returns `True` if value is NOT present in the collection | `x not in seq` |

### Performance Note:
* For **lists and tuples**, `in` performs a linear scan: $O(n)$ time.
* For **sets and dictionaries**, `in` uses a hash lookup: average $O(1)$ constant time.
* For **dictionaries**, `in` checks against **keys** by default, not values.

```python

greeting = "Hello, World!"
print(f"'World' in greeting : {'World' in greeting}")     # True (substring check)
print(f"'z' not in greeting   : {'z' not in greeting}")       # True

```

```python
fruits = ["apple", "banana", "cherry"]
print(f"'banana' in fruits   : {'banana' in fruits}")       # True
print(f"'orange' in fruits   : {'orange' in fruits}")       # False

```

```python

user = {"name": "Alice", "role": "admin", "age": 30}
print(f"'name' in user       : {'name' in user}")           # True (key exists)
print(f"'Alice' in user      : {'Alice' in user}")          # False (value, not key!)
print(f"'Alice' in user.values() : {'Alice' in user.values()}") # True
```

## 7. Bitwise Operators

Bitwise operators manipulate integers at the individual binary (bit) level.

| Operator | Name | Description | Example |
| :--- | :--- | :--- | :--- |
| `&` | Bitwise AND | Sets each bit to 1 if both bits are 1 | `a & b` |
| `|` | Bitwise OR | Sets each bit to 1 if at least one bit is 1 | `a | b` |
| `^` | Bitwise XOR | Sets each bit to 1 if only one bit is 1 | `a ^ b` |
| `~` | Bitwise NOT | Inverts all bits (formula: `~x == -x - 1`) | `~a` |
| `<<` | Left Shift | Shifts bits left by $n$, multiplying by $2^n$ | `a << n` |
| `>>` | Right Shift | Shifts bits right by $n$, dividing by $2^n$ | `a >> n` |

```python
# 7. Bitwise Operators Example
p = 10  # In binary: 1010
q = 4   # In binary: 0100

print(f"p = {p} (bin: {bin(p)[2:].zfill(4)})")
print(f"q = {q}  (bin: {bin(q)[2:].zfill(4)})")
print("-" * 25)

print(f"p & q  : {p & q}   (bin: {bin(p & q)[2:].zfill(4)})")  # 0000 -> 0
print(f"p | q  : {p | q}  (bin: {bin(p | q)[2:].zfill(4)})")  # 1110 -> 14
print(f"p ^ q  : {p ^ q}  (bin: {bin(p ^ q)[2:].zfill(4)})")  # 1110 -> 14
print(f"~p     : {~p}  (two's complement: -p - 1)")          # -11

print("\n--- Bit Shifting ---")
num = 5  # binary: 0101
print(f"num = {num} (bin: {bin(num)})")
print(f"num << 1 : {num << 1} (bin: {bin(num << 1)}) -> multiplies by 2")
print(f"num << 2 : {num << 2} (bin: {bin(num << 2)}) -> multiplies by 4")
print(f"num >> 1 : {num >> 1} (bin: {bin(num >> 1)}) -> floor divides by 2")
```

## 8. Modern & Special Operators

Python introduces specialized operators for modern programming patterns:

1. **Walrus Operator (`:=`)** (Python 3.8+): Assignment expression. Allows assigning a value to a variable within an expression (e.g., inside an `if` condition or `while` loop).
2. **Matrix Multiplication Operator (`@`)** (Python 3.5+): Used in numerical linear algebra (e.g., NumPy).
3. **Dictionary Merge (`|`) and Update (`|=`)** (Python 3.9+): Merges two dictionaries without calling `.update()` or copying.
4. **Unpacking Operators (`*` and `**`)**:
   * Single asterisk `*`: Iterable unpacking into elements or arguments.
   * Double asterisk `**`: Dictionary / keyword argument unpacking.

```python

# 4. Matrix Multiplication (@)
class Simple2DMatrix:
    def __init__(self, data):
        self.data = data
    def __matmul__(self, other):
        # 2x2 matrix multiplication demonstration
        a, b = self.data, other.data
        return Simple2DMatrix([
            [a[0][0]*b[0][0] + a[0][1]*b[1][0], a[0][0]*b[0][1] + a[0][1]*b[1][1]],
            [a[1][0]*b[0][0] + a[1][1]*b[1][0], a[1][0]*b[0][1] + a[1][1]*b[1][1]]
        ])
    def __repr__(self):
        return f"{self.data[0]}\n{self.data[1]}"

m1 = Simple2DMatrix([[1, 2], [3, 4]])
m2 = Simple2DMatrix([[5, 6], [7, 8]])
print(f"\n Matrix Multiplication (@):\n{m1 @ m2}")
```

## 9. Operator Precedence and Associativity

**Precedence** dictates the order in which Python evaluates different operators in an expression. **Associativity** dictates evaluation order when operators have the same precedence.

### Precedence Table (Highest to Lowest):
1. `( )` Parentheses (Grouping)
2. `**` Exponentiation (*Right-to-Left associative*)
3. `+x`, `-x`, `~x` Unary positive, negative, bitwise NOT
4. `*`, `/`, `//`, `%` Multiplication, True division, Floor division, Modulus
5. `+`, `-` Addition, Subtraction
6. `<<`, `>>` Bitwise Shifts
7. `&` Bitwise AND
8. `^` Bitwise XOR
9. `|` Bitwise OR
10. `==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `is not`, `in`, `not in` Comparisons & Memberships
11. `not` Logical NOT
12. `and` Logical AND
13. `or` Logical OR
14. `:=` Assignment expression (Walrus)

> **Best Practice**: Whenever an expression is non-trivial, use **parentheses `()`** to make intent explicit and avoid subtle bugs.

```python
# 9. Operator Precedence and Associativity Example

# Exponentiation has right-to-left associativity:
# 2 ** 3 ** 2 is evaluated as 2 ** (3 ** 2) = 2 ** 9 = 512 (NOT (2**3)**2 = 64)
result_assoc = 2 ** 3 ** 2
print(f"2 ** 3 ** 2 = {result_assoc}")
print(f"(2 ** 3) ** 2 = {(2 ** 3) ** 2}")

# Arithmetic vs Comparison vs Logical
# 10 + 2 * 3 > 15 and not 4 > 5
# Evaluated as:
# 1. 2 * 3 = 6
# 2. 10 + 6 = 16
# 3. 16 > 15 -> True
# 4. 4 > 5 -> False
# 5. not False -> True
# 6. True and True -> True
complex_expr = 10 + 2 * 3 > 15 and not 4 > 5
print(f"\nComplex expression evaluated: {complex_expr}")

# Parentheses ensure clarity
clarified_expr = ((10 + (2 * 3)) > 15) and (not (4 > 5))
print(f"Explicitly parenthesized:    {clarified_expr}")
```

## 10. Operator Overloading & The `operator` Module

Python allows custom classes to define how operators interact with them using **special / magic methods** (also known as **dunder methods**).

### Common Operator Dunders:
| Operator | Dunder Method | Reverse Method | In-Place Method |
| :--- | :--- | :--- | :--- |
| `+` | `__add__(self, other)` | `__radd__` | `__iadd__` |
| `-` | `__sub__(self, other)` | `__rsub__` | `__isub__` |
| `*` | `__mul__(self, other)` | `__rmul__` | `__imul__` |
| `/` | `__truediv__(self, other)` | `__rtruediv__` | `__itruediv__` |
| `==` | `__eq__(self, other)` | - | - |
| `<` | `__lt__(self, other)` | - | - |

### The `operator` Standard Library Module:
The `operator` module exports efficient C-implemented functions corresponding to the intrinsic operators of Python (e.g. `operator.add(x, y)` == `x + y`), often used in functional tools like `functools.reduce` or `sorted()`.

```python
# 10. Operator Overloading & The operator Module Example
import operator
from functools import reduce

# Custom 2D Vector class with overloaded operators
class Vector2D:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    # Overload '+' operator
    def __add__(self, other: "Vector2D"):
        if isinstance(other, Vector2D):
            return Vector2D(self.x + other.x, self.y + other.y)
        return NotImplemented

    # Overload '*' operator (scalar multiplication)
    def __mul__(self, scalar: float):
        if isinstance(scalar, (int, float)):
            return Vector2D(self.x * scalar, self.y * scalar)
        return NotImplemented

    # Overload '==' operator
    def __eq__(self, other: object):
        if isinstance(other, Vector2D):
            return self.x == other.x and self.y == other.y
        return False

    def __repr__(self):
        return f"Vector2D(x={self.x}, y={self.y})"

# Test custom overloaded operators
v1 = Vector2D(2, 5)
v2 = Vector2D(3, 1)

print("--- Custom Operator Overloading ---")
print(f"v1 + v2     : {v1 + v2}")       # Calls v1.__add__(v2)
print(f"v1 * 3      : {v1 * 3}")        # Calls v1.__mul__(3)
print(f"v1 == v2    : {v1 == v2}")       # Calls v1.__eq__(v2)
print(f"v1 == Vector2D(2, 5): {v1 == Vector2D(2, 5)}")

print("\n--- Using the built-in 'operator' module ---")
print(f"operator.add(10, 5) : {operator.add(10, 5)}")
print(f"operator.mul(4, 7)  : {operator.mul(4, 7)}")

# Practical use with reduce: calculate product of a list
numbers = [1, 2, 3, 4, 5]
product = reduce(operator.mul, numbers)
print(f"Product of {numbers} using reduce & operator.mul: {product}")
```

```python

# Sorting items by key using operator.itemgetter
students = [("Alice", 88), ("Bob", 72), ("Charlie", 95)]
sorted_students = sorted(students, key=operator.itemgetter(1), reverse=True)
print(f"Students sorted by score: {sorted_students}")
```
