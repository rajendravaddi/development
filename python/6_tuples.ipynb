## 1. Definition and Creation

A **tuple** in Python is an **immutable, ordered collection** of elements. Once created, elements cannot be added, removed, or changed.

### Creation Ways:
* Parentheses: `(1, 2, 3)`
* Trailing comma without parentheses (Tuple packing): `1, 2, 3`
* Empty tuple: `()` or `tuple()`
* Constructor: `tuple(iterable)` (from strings, lists, sets, ranges)

````python
t_empty = ()
print("Empty tuple :", t_empty)
````

````python
t_nums = (10, 20, 30, 40)
print("Numbers     :", t_nums)
````

````python
t_mixed = (1, "Python", 3.14, True, None)
print("Mixed types :", t_mixed)
````

````python
t_packed = 100, 200, 300
print("Packed tuple:", t_packed, type(t_packed))
````

````python
t_from_list = tuple([1, 2, 3])
print("From list   :", t_from_list)
````

````python
t_from_str = tuple("ABC")

print("From string :", t_from_str)
````

## 2. The Singleton Tuple Quirk (Trailing Comma)

A common mistake in Python: `(42)` is **not** a tuple; parentheses in Python are used for mathematical grouping, so `(42)` evaluates to the integer `42`.

To create a single-element tuple, you **must include a trailing comma**:
* `(42,)` or `42,`

````python
not_a_tuple = (42)
print(f"(42)   type: {type(not_a_tuple)} -> Value: {not_a_tuple}")
````

````python
is_a_tuple = (42,)
print(f"(42,)  type: {type(is_a_tuple)} -> Value: {is_a_tuple}")
````

````python
also_a_tuple = 42,
print(f"42,    type: {type(also_a_tuple)} -> Value: {also_a_tuple}")
````

## 3. Memory Management & Optimization (Tuples vs Lists)

### Why Tuples Are Faster and More Memory-Efficient than Lists:
1. **Fixed Memory Allocation**: Since tuples are immutable, Python allocates the exact size needed without any spare capacity (no over-allocation).
2. **Tuple Free List / PyTupleObject Caching**: CPython maintains a cache of small recycled tuples (up to length 20). Reallocating a 1-tuple or 2-tuple avoids calling the system heap allocator.
3. **Execution Speed**: Iterating and instantiating tuples is measurably faster than lists.

````python
# 3. Memory and Speed Comparison Example
import sys
import timeit

lst = [1, 2, 3, 4, 5]
tpl = (1, 2, 3, 4, 5)

print(f"List size in RAM : {sys.getsizeof(lst)} bytes")
print(f"Tuple size in RAM: {sys.getsizeof(tpl)} bytes (More compact!)")

# Creation speed comparison
list_time = timeit.timeit("[1, 2, 3, 4, 5]", number=1_000_000)
tuple_time = timeit.timeit("(1, 2, 3, 4, 5)", number=1_000_000)

print(f"\nTime to instantiate 1M lists : {list_time:.4f}s")
print(f"Time to instantiate 1M tuples: {tuple_time:.4f}s (Tuples are faster!)")
````

## 4. Immutability & The Mutable-Element Nuance

Tuples are immutable: you cannot reassign items, append, or delete items.

> **Crucial Nuance**: If a tuple contains a **mutable element** (such as a `list` or `dict`), the contents of that child object **can** still be mutated!
> Only the tuple's *references* are fixed.

````python
t = (1, 2, [10, 20])

# Direct reassignment raises TypeError
try:
    t[0] = 99
except TypeError as e:
    print("Caught expected TypeError:", e)
````

````python
t = (1, 2, [10, 20])

# Modifying the mutable list inside the tuple
print("\nBefore modifying child list:", t)
t[2].append(30)
t[2][0] = 999
print("After modifying child list :", t)
````

````python
t = (1, 2, [10, 20])

# Notice: Hashability check
# A tuple is only hashable (usable as a dict key or set element)
# if ALL of its elements are themselves hashable/immutable:
try:
    hash(t)
except TypeError as e:
    print("Cannot hash tuple containing a list:", e)
````

````python
pure_tuple = (1, 2, 3)
print("Hash of pure tuple:", hash(pure_tuple))
````

## 5. Tuple Packing & Unpacking

Tuple packing and unpacking is an idiomatic Python feature:
* **Multiple assignment**: `x, y = 10, 20`
* **Swapping variables without a temporary variable**: `a, b = b, a`
* **Extended unpacking with `*`**: Captures remaining items into a list.

````python
# 1. Basic unpacking
coords = (37.7749, -122.4194)
lat, lon = coords
print(f"Latitude: {lat}, Longitude: {lon}")
````

````python
# 2. Variable swap
x, y = 5, 10
x, y = y, x
print(f"Swapped: x={x}, y={y}")
````

````python
# 3. Extended unpacking with *
person_record = ("Alice", "Engineer", 29, "New York", "USA")
name, profession, age, *location = person_record
print(f"Name: {name}, Age: {age}, Location: {location}")
````

````python
# 4. In loops (enumerate, dict.items())
pairs = [(1, "one"), (2, "two"), (3, "three")]
for number, word in pairs:
    print(f" {number} -> {word}")
````

## 6. Indexing, Slicing & Operations

Tuples support all standard sequence operations that do not modify data:
* Indexing `t[0]`, `t[-1]`
* Slicing `t[1:3]`, `t[::-1]`
* Membership `x in t`, `x not in t`
* Concatenation `+` and repetition `*` (returns a new tuple)

````python
t = (10, 20, 30, 40, 50, 60)

print("First item :", t[0])
print("Last item  :", t[-1])
print("Slice [1:4]:", t[1:4])
print("Reversed   :", t[::-1])
````

````python
t = (10, 20, 30, 40, 50, 60)
# Membership
print("30 in t    :", 30 in t)
print("99 not in t:", 99 not in t)
````

````python

# Operations
t1 = (1, 2)
t2 = (3, 4)
print("Concatenation t1 + t2:", t1 + t2)
print("Repetition t1 * 3    :", t1 * 3)
````

## 7. All Built-in Tuple Methods

Because tuples are immutable, they provide only two built-in methods: `count()` and `index()`.

### `count()`

**Functionality**: Returns the number of times a specified value occurs in the tuple.

**Syntax**:
```python
tuple.count(value) -> int
```

````python
# count() example
t = (1, 2, 3, 2, 4, 2, 5)

print("Count of 2:", t.count(2))
print("Count of 9:", t.count(9))
````

### `index()`

**Functionality**: Returns the lowest index of the specified value. Raises ValueError if the value is not present.

**Syntax**:
```python
tuple.index(value[, start[, stop]]) -> int
```

````python
# index() example
letters = ("a", "b", "c", "b", "d")

first_b = letters.index("b")
print("First 'b' at index       :", first_b)

second_b = letters.index("b", first_b + 1)
print("Second 'b' starting at 2 :", second_b)

try:
    letters.index("z")
except ValueError as e:
    print("Caught error:", e)
````

## 8. Named Tuples (`namedtuple` and `NamedTuple`)

Named tuples combine the memory efficiency and immutability of tuples with readable, dot-notation field access (like a lightweight class).

1. `collections.namedtuple`: Factory function.
2. `typing.NamedTuple`: Class-based syntax with type hints.

````python
# 1. Using collections.namedtuple
from collections import namedtuple
Point = namedtuple("Point", ["x", "y"])
p1 = Point(10, 20)
print("namedtuple p1       :", p1)
print("Access by attribute : p1.x =", p1.x, "p1.y =", p1.y)
print("Access by index     : p1[0] =", p1[0])
````

````python
# 2. Using typing.NamedTuple (with type hints)
from typing import NamedTuple
class Employee(NamedTuple):
    name: str
    role: str
    salary: float

emp = Employee("Alice", "Lead Architect", 150000.0)
print("\ntyping.NamedTuple emp:", emp)
print(f"{emp.name} is a {emp.role} earning ${emp.salary:,.2f}")
````

## 9. When to Use Tuples vs Lists (Summary)

| Criteria | Tuple | List |
| :--- | :--- | :--- |
| **Mutability** | Immutable | Mutable |
| **Typical Data** | Heterogeneous records (e.g. coordinates `(x, y)`, DB row) | Homogeneous collections (e.g. `[user1, user2]`) |
| **Dictionary Keys** | Allowed (if all elements are immutable) | Not allowed (`TypeError: unhashable type`) |
| **Memory & Performance** | Smaller memory, faster creation | Dynamic resizing overhead |
| **Methods** | Only 2 (`count`, `index`) | 11 mutating & query methods |
