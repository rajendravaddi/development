## 1. Definition and Creation

A **set** in Python is an **unordered, mutable collection of unique, hashable elements**.
* **Unique**: Duplicate elements are automatically discarded.
* **Unordered**: Elements do not maintain index positions (indexing `s[0]` is not supported).
* **Hashable elements**: Set items must be immutable/hashable.

### How to Create Sets:
* Curly braces: `{1, 2, 3}`
* Constructor: `set(iterable)` (e.g. from lists or strings to deduplicate)
* **Empty set**: Must be created with `set()`, because `{}` creates an empty **dictionary**!

````python
# 1. Set Creation Examples
# Deduplication of elements
numbers = {1, 2, 2, 3, 4, 4, 4, 5}
print("Set with duplicates removed:", numbers)

# Empty set check
empty_dict = {}
empty_set = set()
print("type({})     :", type(empty_dict))  # <class 'dict'>
print("type(set())  :", type(empty_set))   # <class 'set'>

# From string or list
chars = set("mississippi")
print("Unique characters in 'mississippi':", chars)
````

## 2. Hash Table Implementation & $O(1)$ Performance

Like dictionaries, sets in Python are powered by **hash tables** (storing keys with no values).
* Checking membership (`x in my_set`) operates in **average $O(1)$ constant time**, regardless of whether the set has 10 elements or 10,000,000 elements!
* In contrast, checking membership in a list (`x in my_list`) is $O(n)$ linear scan.

````python
# 2. Performance Comparison: Set vs List Lookup
import time

large_list = list(range(1_000_000))
large_set = set(range(1_000_000))

# Search for element at the very end
search_target = 999_999

# Measure list membership check
t0 = time.perf_counter()
_ = search_target in large_list
list_duration = time.perf_counter() - t0

# Measure set membership check
t0 = time.perf_counter()
_ = search_target in large_set
set_duration = time.perf_counter() - t0

print(f"List search time (O(n)): {list_duration:.6f}s")
print(f"Set search time  (O(1)): {set_duration:.6f}s")
print(f"Set was {list_duration / set_duration:.1f}x faster!")
````

## 3. Mathematical Set Operations

Python provides mathematical operations both via **operators** and **methods**:

| Mathematical Concept | Operator | Method Equivalent | Description |
| :--- | :--- | :--- | :--- |
| **Union** | `A \| B` | `A.union(B)` | Elements in either A or B or both |
| **Intersection** | `A & B` | `A.intersection(B)` | Elements in both A and B |
| **Difference** | `A - B` | `A.difference(B)` | Elements in A but not in B |
| **Symmetric Difference** | `A ^ B` | `A.symmetric_difference(B)` | Elements in A or B, but not both |
| **Subset** | `A <= B` | `A.issubset(B)` | Is every element of A in B? |
| **Superset** | `A >= B` | `A.issuperset(B)` | Does A contain every element of B? |
| **Disjoint** | - | `A.isdisjoint(B)` | Do A and B share zero common elements? |

````python
# 3. Mathematical Operations Example
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(f"A = {a}")
print(f"B = {b}\n")

print("Union (A | B)               :", a | b)
print("Intersection (A & B)        :", a & b)
print("Difference (A - B)          :", a - b)
print("Difference (B - A)          :", b - a)
print("Symmetric Difference (A ^ B):", a ^ b)
````

---
## 4. Crucial Difference: Operators vs Methods

* **Operators (`|`, `&`, `-`, `^`)**: Require **both** operands to be `set` objects.
* **Methods (`union()`, etc.)**: Accept **any iterable** (list, tuple, generator, dict) as arguments!

````python
# 4. Operator vs Method Flexibility Example
my_set = {1, 2, 3}
my_list = [3, 4, 5]

# Method works seamlessly with a list:
result_method = my_set.union(my_list)
print("my_set.union(my_list):", result_method)

# Operator with a list raises TypeError:
try:
    _ = my_set | my_list
except TypeError as e:
    print("Caught TypeError using '|' with list:", e)
````

## 5. Set Comprehensions

Syntax:
```python
{expression for item in iterable if condition}
```

````python
# 5. Set Comprehension Example
# Extract unique word lengths
quote = "the quick brown fox jumps over the lazy dog"
word_lengths = {len(word) for word in quote.split()}
print("Unique word lengths:", word_lengths)

# Set of even squares
even_squares = {x ** 2 for x in range(10) if x % 2 == 0}
print("Even squares:", even_squares)
````

## 6. All Built-in Set Methods (One by One)

Python provides 17 built-in methods on `set` objects. Each is detailed below with its functionality, syntax, and an executable code cell.

### `add()`

**Functionality**: Adds an element to the set. If the element is already present, has no effect.

**Syntax**:
```python
set.add(element) -> None
```

````python
# add() example
fruits = {"apple", "banana"}
fruits.add("cherry")
print("After adding 'cherry':", fruits)

fruits.add("apple")  # Duplicate has no effect
print("After adding 'apple' :", fruits)
````

### `remove()`

**Functionality**: Removes the specified element from the set in-place. Raises KeyError if element is not found.

**Syntax**:
```python
set.remove(element) -> None
```

````python
# remove() example
colors = {"red", "green", "blue"}
colors.remove("green")
print("After remove('green'):", colors)

try:
    colors.remove("yellow")
except KeyError as e:
    print("Caught KeyError:", e)
````

### `discard()`

**Functionality**: Removes the specified element from the set if present. Does NOT raise an error if element is absent.

**Syntax**:
```python
set.discard(element) -> None
```

````python
# discard() example
colors = {"red", "green", "blue"}
colors.discard("blue")
print("After discard('blue')  :", colors)

# Safe: Does not raise KeyError
colors.discard("yellow")
print("After discard('yellow'):", colors)
````

### `pop()`

**Functionality**: Removes and returns an arbitrary element from the set. Raises KeyError if the set is empty.

**Syntax**:
```python
set.pop() -> element
```

````python
# pop() example
items = {"a", "b", "c", "d"}
removed_item = items.pop()
print(f"Popped item: '{removed_item}', Remaining set: {items}")
````

### `clear()`

**Functionality**: Removes all elements from the set in-place, leaving it empty.

**Syntax**:
```python
set.clear() -> None
```

````python
# clear() example
s = {1, 2, 3, 4}
s.clear()
print("After clear:", s)
````

### `copy()`

**Functionality**: Returns a shallow copy of the set.

**Syntax**:
```python
set.copy() -> set
```

````python
# copy() example
orig = {10, 20, 30}
cpy = orig.copy()
cpy.add(40)

print("Original:", orig)
print("Copied  :", cpy)
````

### `union()`

**Functionality**: Returns a new set containing all distinct elements from this set and all specified iterables.

**Syntax**:
```python
set.union(*others) -> set
```

````python
# union() example
s1 = {1, 2}
s2 = {2, 3}
s3 = [3, 4, 5]

print("Union of multiple iterables:", s1.union(s2, s3))
````

### `update()`

**Functionality**: Updates the set in-place, adding all elements from all specified iterables. Equivalent to `|=`.

**Syntax**:
```python
set.update(*others) -> None
```

````python
# update() example
nums = {1, 2}
nums.update([3, 4], {5, 6})
print("After update:", nums)
````

### `intersection()`

**Functionality**: Returns a new set with elements common to this set and all specified iterables.

**Syntax**:
```python
set.intersection(*others) -> set
```

````python
# intersection() example
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
c = (4, 5, 6, 7)

print("Intersection:", a.intersection(b, c))  # {4}
````

### `intersection_update()`

**Functionality**: Updates the set in-place, keeping only elements found in both this set and all specified iterables. Equivalent to `&=`.

**Syntax**:
```python
set.intersection_update(*others) -> None
```

````python
# intersection_update() example
a = {1, 2, 3, 4}
a.intersection_update({3, 4, 5})
print("After intersection_update:", a)
````

### `difference()`

**Functionality**: Returns a new set with elements in this set that are not in the specified iterables.

**Syntax**:
```python
set.difference(*others) -> set
```

````python
# difference() example
a = {1, 2, 3, 4, 5}
b = {4, 5, 6}

print("A difference B:", a.difference(b))  # {1, 2, 3}
````

### `difference_update()`

**Functionality**: Removes all elements of another set from this set in-place. Equivalent to `-=`.

**Syntax**:
```python
set.difference_update(*others) -> None
```

````python
# difference_update() example
a = {1, 2, 3, 4, 5}
a.difference_update([4, 5])
print("After difference_update:", a)  # {1, 2, 3}
````

### `symmetric_difference()`

**Functionality**: Returns a new set with elements in either this set or other, but not in both.

**Syntax**:
```python
set.symmetric_difference(other) -> set
```

````python
# symmetric_difference() example
a = {1, 2, 3}
b = {3, 4, 5}

print("Symmetric difference:", a.symmetric_difference(b))  # {1, 2, 4, 5}
````

### `symmetric_difference_update()`

**Functionality**: Updates this set in-place with the symmetric difference of itself and other. Equivalent to `^=`.

**Syntax**:
```python
set.symmetric_difference_update(other) -> None
```

````python
# symmetric_difference_update() example
a = {1, 2, 3}
a.symmetric_difference_update({3, 4, 5})
print("After symmetric_difference_update:", a)
````

### `issubset()`

**Functionality**: Returns True if every element in this set is present in other.

**Syntax**:
```python
set.issubset(other) -> bool
```

````python
# issubset() example
small = {1, 2}
big = {1, 2, 3, 4}

print("small.issubset(big):", small.issubset(big))  # True
print("big.issubset(small):", big.issubset(small))  # False
````

### `issuperset()`

**Functionality**: Returns True if this set contains every element present in other.

**Syntax**:
```python
set.issuperset(other) -> bool
```

````python
# issuperset() example
big = {1, 2, 3, 4}
small = {1, 2}

print("big.issuperset(small):", big.issuperset(small))  # True
````

### `isdisjoint()`

**Functionality**: Returns True if this set has no elements in common with other (intersection is empty).

**Syntax**:
```python
set.isdisjoint(other) -> bool
```

````python
# isdisjoint() example
a = {1, 2, 3}
b = {4, 5, 6}
c = {3, 7, 8}

print("a.isdisjoint(b):", a.isdisjoint(b))  # True (no common elements)
print("a.isdisjoint(c):", a.isdisjoint(c))  # False (share element 3)
````

## 7. Frozenset (Immutable Sets)

A `frozenset` is an **immutable version of a set**:
* Once created, elements cannot be added or removed.
* Because it is immutable, it is **hashable**, meaning a `frozenset` can be used as a **dictionary key** or as an element inside another **set**!

````python
# 7. Frozenset Examples
fs = frozenset([1, 2, 3, 4])
print("Frozenset:", fs)

# Attempting to mutate raises AttributeError
try:
    fs.add(5)
except AttributeError as e:
    print("Caught error:", e)

# Frozenset as a dictionary key:
access_permissions = {
    frozenset(["read", "write"]): "Editor",
    frozenset(["read"]): "Viewer",
    frozenset(["read", "write", "admin"]): "Owner"
}

user_perms = frozenset(["read", "write"])
print("Role for permissions:", access_permissions[user_perms])

# Frozensets inside a regular set (Set of sets)
set_of_sets = {frozenset([1, 2]), frozenset([3, 4])}
print("Set of frozensets:", set_of_sets)
````

## 8. Set Complexity & Summary

| Operation | Average Complexity | Worst Case |
| :--- | :--- | :--- |
| Add element (`add`) | $O(1)$ | $O(n)$ |
| Remove element (`remove` / `discard`) | $O(1)$ | $O(n)$ |
| Membership test (`in`) | $O(1)$ | $O(n)$ |
| Pop element (`pop`) | $O(1)$ | $O(1)$ |
| Union (`A \| B`) | $O(\text{len}(A) + \text{len}(B))$ | $O(\text{len}(A) + \text{len}(B))$ |
| Intersection (`A & B`) | $O(\min(\text{len}(A), \text{len}(B)))$ | $O(\text{len}(A) \cdot \text{len}(B))$ |
| Difference (`A - B`) | $O(\text{len}(A))$ | $O(\text{len}(A) \cdot \text{len}(B))$ |

````python
# 8. Practical Deduplication Idiom
# Fast deduplication preserving or not preserving order:
raw_list = ["apple", "banana", "apple", "cherry", "banana"]

# Unordered deduplication
deduped_set = list(set(raw_list))
print("Unordered deduplication :", deduped_set)

# Order-preserving deduplication using dict.fromkeys (Python 3.7+)
deduped_ordered = list(dict.fromkeys(raw_list))
print("Order-preserving dedup  :", deduped_ordered)
````
