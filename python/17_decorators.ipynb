# Python Decorators

Decorators are one of Python's most elegant and powerful design patterns. They allow developers to modify, enhance, or extend the behavior of functions or methods without permanently altering their underlying source code.

### 1. What Are Decorators?

A **decorator** is a design pattern in Python used to extend or modify the behavior of a function or method without altering its original code. Under the hood, a decorator is simply a callable (usually a higher-order function) that takes another function as an argument, adds extra functionality around it, and returns the modified or replacement callable.

**Syntax**:
```python
@my_decorator
def target_function():
    # original logic
```

````python
# 1. Conceptual Decorator Example
def simple_decorator(func):
    def wrapper():
        print("--> Before the function executes")
        func()
        print("--> After the function executes")
    return wrapper

@simple_decorator
def say_hello():
    print("Hello, world!")

say_hello()
````

### 2. First-Class Functions in Python

In Python, functions are **first-class citizens** (or first-class objects). This means functions can be:
- Assigned to variables
- Passed as arguments to other functions
- Returned from other functions
- Stored in data structures (lists, dictionaries, etc.)

**Syntax**:
```python
f = my_func
my_list = [func_a, func_b]
```

````python
# 2. Treating functions as objects
def shout(text):
    return text.upper() + "!"

# Assigning a function to a variable
yell = shout

print("Original function call :", shout("hello"))
print("Assigned variable call :", yell("hello"))
print("Function object name   :", yell.__name__)

````

````python
# Storing functions in a list and calling them dynamically
operations = [str.lower, str.upper, str.title]
sample_string = "python programming"
for op in operations:
    print(f"{op.__name__:<10}: {op(sample_string)}")
````

### 3. Functions as Arguments

Because functions are first-class citizens, a function can accept another function as a parameter. These are known as **higher-order functions**.

**Syntax**:
```python
def apply_func(func, value):
    return func(value)
```

````python
# 3. Passing functions as arguments
def square(x):
    return x * x

def cube(x):
    return x * x * x

def apply_transformation(func, numbers):
    """Higher-order function applying 'func' to each item in 'numbers'."""
    return [func(n) for n in numbers]

nums = [1, 2, 3, 4]
print("Squares:", apply_transformation(square, nums))
print("Cubes  :", apply_transformation(cube, nums))
````

### 4. Functions Returning Other Functions

A function can generate and return another function object dynamically. This is a crucial prerequisite for building decorators and closures.

**Syntax**:
```python
def factory():
    def new_func(): ...
    return new_func
```

````python
# 4. Function generating and returning another function
def greeting_maker(greeting_word):
    def greet(name):
        return f"{greeting_word}, {name}!"
    return greet

# Create specialized greeting functions
say_hi = greeting_maker("Hi")
say_hola = greeting_maker("Hola")

print(say_hi("Alice"))
print(say_hola("Carlos"))
````

### 5. Nested Functions (Inner Functions)

A function defined inside the body of another function is called a **nested** (or inner) function. Inner functions are scoped locally to the outer enclosing function and are not accessible from the global scope unless returned.

**Syntax**:
```python
def outer():
    def inner():
        ...
    return inner()
```

````python
# 5. Nested functions demonstration
def outer_function(message):
    print("Executing outer_function")
    
    def inner_function():
        print(f"Executing inner_function with message: '{message}'")
    
    # Call the inner function internally
    inner_function()

outer_function("Local context preserved")

# Notice: Calling inner_function directly from outside raises NameError
try:
    inner_function()
except NameError as e:
    print("Caught expected error:", e)
````

### 6. Closures in Python

A **closure** is a nested function that remembers and retains access to variables in its enclosing scope (lexical scope), even after the outer function has finished executing and exited the stack.

A closure requires: a nested function, referencing a variable from the enclosing function, and the outer function returning the nested function.

**Syntax**:
```python
def make_multiplier(n):
    def multiplier(x):
        return x * n  # 'n' is enclosed / captured
    return multiplier
```

````python
# 6. Closures and inspecting cell contents
def counter_factory(start=0):
    count = start  # Enclosed variable
    
    def increment(step=1):
        nonlocal count
        count += step
        return count
    
    return increment

my_counter = counter_factory(100)
print("Call 1:", my_counter())    # 101
print("Call 2:", my_counter(5))   # 106
print("Call 3:", my_counter(10))  # 116

# Inspect the captured closure variable
print("Closure cell contents:", my_counter.__closure__[0].cell_contents)
````

## 7. Understanding the `@` Syntax (Syntactic Sugar)

The `@decorator_name` syntax placed above a function definition is simply convenient **syntactic sugar** introduced in PEP 318.

Writing:
```python
@my_decorator
def greet():
    print("Hello")
```

Is 100% equivalent to:
```python
def greet():
    print("Hello")

# Rebinding the function name to the decorated wrapper
greet = my_decorator(greet)
```

````python
# 7. Demonstrating syntactic sugar equivalence
def announcement_decorator(func):
    def wrapper():
        print("[ANNOUNCEMENT] Starting operation...")
        func()
        print("[ANNOUNCEMENT] Operation concluded.")
    return wrapper

# Style A: Traditional manual reassignment
def task_a():
    print("Executing Task A")

task_a = announcement_decorator(task_a)

# Style B: Using the clean @ syntax
@announcement_decorator
def task_b():
    print("Executing Task B")

print("--- Running Task A (Manual reassignment) ---")
task_a()

print("\n--- Running Task B (Using @ syntax) ---")
task_b()
````

### 8. Creating Your First Decorator

To build a standard decorator: define an outer function accepting `func`, define an inner `wrapper` function that adds before/after logic and calls `func()`, and return `wrapper`.

**Syntax**:
```python
def my_decorator(func):
    def wrapper():
        # before
        func()
        # after
    return wrapper
```

````python
# 8. Building a banner / border decorator
def banner_decorator(func):
    def wrapper():
        print("=" * 35)
        func()
        print("=" * 35)
    return wrapper

@banner_decorator
def print_invoice_header():
    print("OFFICIAL INVOICE 2026".center(35))

print_invoice_header()
````

### 9. How Decorators Work Internally

When Python executes the `@decorator` line at definition time (compile/import time, NOT when the function is called!), it passes the original function into the decorator and rebinds the function's name to the returned wrapper function.


````python
# 9. Verifying definition-time execution
def trace_decorator(func):
    print(f"** [TRACE] Decorator executed at definition time for: {func.__name__} **")
    def wrapper():
        print(f"** [TRACE] Wrapper executed at call time for: {func.__name__} **")
        return func()
    return wrapper

print("Defining decorated function...")
@trace_decorator
def compute_data():
    print("Computing data...")

print("Function defined! Notice the decorator has already executed.")
print("\nNow calling compute_data() twice:")
compute_data()
compute_data()
````

### 10. Decorators with Arguments (`*args`, `**kwargs`)

If the decorated function takes parameters, the inner `wrapper` function must accept `*args` and `**kwargs` and pass them along to the original function. This ensures the decorator works on any function regardless of its signature.

**Syntax**:
```python
def decorator(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper
```

````python
# 10. Universal decorator accepting arbitrary arguments
def audit_arguments(func):
    def wrapper(*args, **kwargs):
        print(f"[AUDIT] Calling {func.__name__}() with:")
        print(f"        Positional args: {args}")
        print(f"        Keyword args   : {kwargs}")
        return func(*args, **kwargs)
    return wrapper

@audit_arguments
def add_user(username, role="guest", active=True):
    print(f"User '{username}' registered with role '{role}' (Active: {active})")

add_user("Alice", role="Admin")
print()
add_user("Bob", "Developer", active=False)
````

### 11. Returning Values from Decorators

A common beginner mistake is forgetting to return the result of `func(*args, **kwargs)` inside the wrapper. If you omit the return statement, the decorated function will return `None` regardless of what the original function computed.

**Syntax**:
```python
def decorator(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result  # Crucial!
    return wrapper
```

````python
# 11. Preserving return values
def double_result_decorator(func):
    def wrapper(*args, **kwargs):
        raw_result = func(*args, **kwargs)
        # Double the return value
        return raw_result * 2
    return wrapper

@double_result_decorator
def multiply(a, b):
    return a * b

output = multiply(5, 4)  # (5 * 4) * 2 = 40
print("Result with return preserved and modified:", output)
````

### 12. Preserving Metadata with `functools.wraps`

Decorating a function replaces it with the `wrapper` function, which accidentally overwrites the original function's `__name__`, `__doc__`, and `__annotations__`. Using `@functools.wraps(func)` on the wrapper automatically copies and preserves all original metadata.

**Syntax**:
```python
from functools import wraps

def decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper
```

````python
# 12. functools.wraps demonstration
from functools import wraps

# BAD: Decorator without wraps
def bad_decorator(func):
    def wrapper(*args, **kwargs):
        """Wrapper docstring."""
        return func(*args, **kwargs)
    return wrapper

@bad_decorator
def calculate_tax(income: float):
    """Calculate annual state income tax."""
    return income * 0.15

print("--- Without functools.wraps ---")
print("Name     :", calculate_tax.__name__)      # 'wrapper' (LOST!)
print("Docstring:", calculate_tax.__doc__)       # 'Wrapper docstring' (LOST!)
````

````python


from functools import wraps

# GOOD: Decorator with wraps
def good_decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper


@good_decorator
def calculate_vat(price: float):
    """Calculate standard value added tax."""
    return price * 0.20


print("\n--- With functools.wraps ---")
print("Name     :", calculate_vat.__name__)      # 'calculate_vat' (PRESERVED!)
print("Docstring:", calculate_vat.__doc__)       # 'Calculate standard value added tax.' (PRESERVED!)
````

### 13. Decorators with Parameters (Decorator Factories)

When a decorator needs arguments of its own (e.g., `@repeat(num_times=3)` or `@route('/home')`), you need three levels of nested functions: the outermost function acts as a **decorator factory** accepting the configuration parameters, the middle function accepts the target `func`, and the innermost `wrapper` handles the execution.

**Syntax**:
```python
def repeat(num_times):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

````python
# 13. Decorator Factory: Repeat execution N times
from functools import wraps

def repeat(num_times: int):
    """Decorator factory that repeats function execution 'num_times'."""
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            last_result = None
            for i in range(1, num_times + 1):
                print(f"[Iteration {i}/{num_times}]")
                last_result = func(*args, **kwargs)
            return last_result
        return wrapper
    return decorator

@repeat(num_times=3)
def send_ping(server_ip):
    print(f"Ping sent to {server_ip}")

send_ping("192.168.1.1")
# print(send_ping.__name__)
# print(send_ping.__doc__)
````

### 14. Multiple Decorators on a Single Function

You can stack multiple decorators on a single function. They are applied from the **bottom-up** (nearest to the function first) and executed from the **top-down** at runtime.

**Syntax**:
```python
@decorator_one
@decorator_two
def func(): ...
# Equivalent to: func = decorator_one(decorator_two(func))
```

````python
# 14. Multiple Decorators Stacking & Execution Order
def make_bold(func):
    def wrapper(*args, **kwargs):
        return f"<b>{func(*args, **kwargs)}</b>"
    return wrapper

def make_italic(func):
    def wrapper(*args, **kwargs):
        return f"<i>{func(*args, **kwargs)}</i>"
    return wrapper

def make_underline(func):
    def wrapper(*args, **kwargs):
        return f"<u>{func(*args, **kwargs)}</u>"
    return wrapper

# Applied bottom-up: underline -> italic -> bold
@make_bold
@make_italic
@make_underline
def format_text(text):
    return text

print("Chained Decorator HTML output:")
print(format_text("Decorators in Python"))
# Output: <b><i><u>Decorators in Python</u></i></b>
````

### 15. Decorators on Classes (Class Decorators)

Decorators can decorate entire classes rather than just functions. A class decorator receives a class object as its argument, inspects or modifies attributes/methods, and returns the modified class.

**Syntax**:
```python
def class_decorator(cls):
    # add attributes or methods
    return cls

@class_decorator
class MyClass: ...
```

````python
# 15. Class decorator: Auto-adding a custom repr and timestamp
from datetime import datetime

def add_audit_timestamp(cls):
    """Class decorator that adds created_at attribute and custom repr to a class."""
    orig_init = cls.__init__
    
    def new_init(self, *args, **kwargs):
        orig_init(self, *args, **kwargs)
        self.created_at = datetime.now()
    
    cls.__init__ = new_init
    cls.audit_info = lambda self: f"{self.__class__.__name__} instance created at {self.created_at}"
    return cls

@add_audit_timestamp
class UserAccount:
    def __init__(self, username):
        self.username = username

user = UserAccount("alice_dev")
print("Username  :", user.username)
print("Audit Info:", user.audit_info())
````

### 16. Class-Based Decorators (`__call__`)

Instead of writing a nested function, a decorator can be implemented as a **class** by defining the special `__call__()` method. This turns instances of the class into callables. Class-based decorators are ideal when the decorator needs to maintain complex state (e.g. execution counts, caches).

**Syntax**:
```python
class ClassDecorator:
    def __init__(self, func):
        self.func = func
    def __call__(self, *args, **kwargs):
        return self.func(*args, **kwargs)
```

````python
# 16. Class-Based Decorator tracking execution counts
class CallCounter:
    def __init__(self, func):
        self.func = func
        self.count = 0  # Persistent state

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"[METRIC] '{self.func.__name__}' has been called {self.count} time(s).")
        return self.func(*args, **kwargs)

@CallCounter
def process_payment(amount):
    print(f"Processing payment of ${amount:,.2f}")

process_payment(150)
process_payment(250)
process_payment(500)
print("Total calls recorded:", process_payment.count)
````

### 17. Decorating Class Methods

Decorating a class method is identical to decorating a regular function, provided the wrapper accepts `*args` and `**kwargs`. Because class methods receive `self` as their first argument, `*args` naturally captures `self` at position 0.

**Syntax**:
```python
class MyClass:
    @my_decorator
    def my_method(self, x): ...
```

````python
# 17. Decorating methods inside a class
from functools import wraps

def log_method_call(func):
    @wraps(func)
    def wrapper(self, *args, **kwargs):
        print(f"[{self.__class__.__name__}.{func.__name__}] invoked with args: {args}")
        return func(self, *args, **kwargs)
    return wrapper

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    @log_method_call
    def deposit(self, amount):
        self.balance += amount
        return self.balance

account = BankAccount("Alice", 1000)
account.deposit(500)
````

### 18. Built-in Decorators: `@property`, `@classmethod`, `@staticmethod`

Python includes three essential built-in decorators for Object-Oriented Programming:
- `@property`: Exposes a method as a getter attribute, with optional `@attr.setter`
- `@classmethod`: Binds the method to the class (`cls`) rather than instance (`self`)
- `@staticmethod`: Defines a utility function bound to the class namespace without access to `self` or `cls`.

**Syntax**:
```python
class MyClass:
    @property
    def value(self): ...
    @classmethod
    def from_string(cls, s): ...
    @staticmethod
    def validate(x): ...
```

````python
# 18. Built-in OOP Decorators Demo
class Temperature:
    def __init__(self, celsius: float):
        self._celsius = celsius

    # 1. @property (Getter)
    @property
    def fahrenheit(self):
        return (self._celsius * 9 / 5) + 32

    # Property Setter
    @fahrenheit.setter
    def fahrenheit(self, value: float):
        self._celsius = (value - 32) * 5 / 9

    # 2. @classmethod (Alternative constructor)
    @classmethod
    def from_fahrenheit(cls, fahrenheit: float):
        celsius = (fahrenheit - 32) * 5 / 9
        return cls(celsius)

    # 3. @staticmethod (Independent utility function)
    @staticmethod
    def is_freezing(celsius: float) -> bool:
        return celsius <= 0.0

# Using @property
t = Temperature(25)
print(f"25°C in Fahrenheit (getter) : {t.fahrenheit}°F")
t.fahrenheit = 212  # setter
print(f"212°F set -> Celsius is now  : {t._celsius}°C")

# Using @classmethod
t_winter = Temperature.from_fahrenheit(14)
print(f"Created from classmethod     : {t_winter._celsius}°C")

# Using @staticmethod
print(f"Is -5°C freezing? (static)   : {Temperature.is_freezing(-5)}")
````

### 19. Practical Real-World Use Cases (Logging, Timing, Auth, Caching)

Decorators are foundational in production frameworks (like Flask, FastAPI, Django, and Celery) for cross-cutting concerns:
1. **Performance profiling / Timing**
2. **Access control / Authentication**
3. **Result caching / Memoization (`functools.lru_cache`)**

**Syntax**:
```python
@timer

@authenticate(role)

@lru_cache
```

````python
# 19. Real-World Production Decorators Examples
import time
from functools import wraps, lru_cache

# 1. Performance Timing Decorator
def measure_time(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        duration = time.perf_counter() - start
        print(f"[TIME] '{func.__name__}' completed in {duration:.6f} seconds.")
        return result
    return wrapper

# 2. Role-Based Authentication Guard Decorator
current_user = {"username": "admin_raj", "role": "admin"}

def require_role(required_role):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            if current_user.get("role") != required_role:
                raise PermissionError(f"Access Denied: Requires '{required_role}' role.")
            return func(*args, **kwargs)
        return wrapper
    return decorator

# 3. Applying Decorators in practice
@measure_time
@require_role("admin")
def delete_database_records():
    print("-> Admin action: Purging outdated records from database...")
    time.sleep(0.05)  # Simulate I/O work
    return "Purge Complete"

print("--- Running Protected & Timed Admin Action ---")
delete_database_records()
````

````python
import time
from functools import wraps, lru_cache

# 4. Built-in Memoization Caching with @lru_cache
@lru_cache(maxsize=128)
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print("\n--- Caching with @lru_cache ---")
t0 = time.perf_counter()
fib_result = fibonacci(35)
t_elapsed = time.perf_counter() - t0
print(f"fibonacci(35) = {fib_result} computed in {t_elapsed:.6f}s (Instant with cache!)")
print("Cache statistics:", fibonacci.cache_info())
````
